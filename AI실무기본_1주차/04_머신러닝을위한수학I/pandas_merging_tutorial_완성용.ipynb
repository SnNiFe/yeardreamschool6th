## 1. `pd.concat()` — 행/열 방향 이어붙이기

> **언제 쓰나?** 같은 구조의 데이터를 단순히 쌓거나 옆에 붙일 때
> (스키마가 동일한 분기별 데이터, 지역별 파일 등)

```python
pd.concat(objs, axis=0, join='outer', ignore_index=False, keys=None)
```

| 파라미터 | 기본값 | 설명 |
|----------|--------|------|
| `axis` | `0` | 0 = 행 방향(세로), 1 = 열 방향(가로) |
| `join` | `'outer'` | `'outer'`(합집합) / `'inner'`(교집합) |
| `ignore_index` | `False` | True면 인덱스를 0, 1, 2 ... 로 초기화 |
| `keys` | `None` | 출처를 나타내는 MultiIndex 레이블 |


````python
# 1-1. axis=0 : 행 방향 (세로로 쌓기)
# 시나리오: Q1 / Q2 주문 데이터를 하나로 합치기

orders_q1 = orders[orders["orderDate"] < "2003-02-01"].copy()
orders_q2 = orders[orders["orderDate"] >= "2003-02-01"].copy()

print(f"Q1 주문: {len(orders_q1)}건  |  Q2 주문: {len(orders_q2)}건")

result = pd.concat([orders_q1, orders_q2], ignore_index=True)
print(f"concat 후 총 행수: {len(result)}건")
result[["orderNumber", "orderDate", "status"]]
````

````python
# 1-2. keys : 출처 레이블 추가 (MultiIndex)
# result.loc["Q1"] 으로 분기별 접근 가능

result_keyed = pd.concat(
    [orders_q1, orders_q2],
    keys=["Q1", "Q2"],
)
print("MultiIndex 구조:")
display(result_keyed[["orderNumber", "orderDate"]].head(7))

# 특정 분기만 추출
print("Q2 주문만 추출:")
result_keyed.loc["Q2"][["orderNumber", "orderDate"]]
````

````python
# 1-3. axis=1 : 열 방향 (가로로 붙이기)
# 시나리오: 고객 기본정보 + 결제 합계를 나란히 합치기
# 인덱스(customerNumber)를 기준으로 정렬되어 붙음

cust_basic = customers[["customerNumber","customerName","country"]].set_index("customerNumber")
pay_total  = (payments
              .groupby("customerNumber")["amount"]
              .sum()
              .rename("totalPaid"))

result_cols = pd.concat([cust_basic, pay_total], axis=1, join="outer")
result_cols
````

````python
# 1-4. join='inner' vs 'outer' 비교
# outer: 양쪽 인덱스의 합집합 -> NaN 발생 가능
# inner: 양쪽 인덱스의 교집합 -> NaN 없음

outer_c = pd.concat([cust_basic, pay_total], axis=1, join="outer")
inner_c = pd.concat([cust_basic, pay_total], axis=1, join="inner")

print(f"outer shape: {outer_c.shape}  (모든 고객 포함)")
print(f"inner shape: {inner_c.shape}  (결제 기록 있는 고객만)")
inner_c
````

## 2. `pd.merge()` — INNER JOIN (기본값)

> **언제 쓰나?** 양쪽 테이블에 **모두 존재하는** 행만 필요할 때

```python
pd.merge(left, right, how='inner', on=None, left_on=None, right_on=None)
```

```
orders        customers
----------    ----------
orderNumber   customerNumber  <- 공통 키
customerNumber
...           customerName
              country
```


````python
# INNER JOIN: orders <-> customers
# customerNumber 가 양쪽에 모두 있는 행만 반환

inner = pd.merge(
    orders,
    customers[["customerNumber","customerName","country","creditLimit"]],
    on="customerNumber",   # 조인 키 (컬럼명이 같을 때)
    how="inner",           # 기본값 -- 생략해도 동일
)

print(f"orders 행수: {len(orders)}  |  customers 행수: {len(customers)}")
print(f"INNER JOIN 결과: {len(inner)}행  (양쪽에 없는 customerNumber 제외)")
inner[["orderNumber","orderDate","customerName","country","creditLimit"]]
````

---
## 3. `pd.merge()` — LEFT JOIN

> **언제 쓰나?** **왼쪽 테이블의 모든 행**을 유지하면서,
> 오른쪽 정보가 없으면 `NaN` 채우기

```
LEFT  : orders    (모두 보존)
RIGHT : customers (매칭되는 것만)
-> 주문이 있어도 customers 에 없는 customerNumber -> NaN
```


````python
left = pd.merge(
    orders,
    customers[["customerNumber","customerName","creditLimit"]],
    on="customerNumber",
    how="left",
)

print(f"orders 행수: {len(orders)}  ->  LEFT JOIN 결과: {len(left)}행  (orders 행수 유지)")
print(f"NaN 개수 (고객 정보 없는 주문):")
print(left[["customerName","creditLimit"]].isna().sum())

left[["orderNumber","orderDate","customerNumber","customerName","creditLimit"]]
````

---
## 4. `pd.merge()` — RIGHT JOIN

> **언제 쓰나?** **오른쪽 테이블의 모든 행**을 유지하면서,
> 왼쪽 정보가 없으면 `NaN` 채우기

```
LEFT  : orders    (매칭되는 것만)
RIGHT : customers (모두 보존)
-> 주문이 한 건도 없는 고객도 결과에 포함
```


````python
right = pd.merge(
    orders[["orderNumber","customerNumber","orderDate","status"]],
    customers[["customerNumber","customerName","country"]],
    on="customerNumber",
    how="right",
)

print(f"customers 행수: {len(customers)}  ->  RIGHT JOIN 결과: {len(right)}행  (customers 행수 유지)")
print(f"\n주문 없는 고객 (orderNumber == NaN):")
display(right[right["orderNumber"].isna()][["customerNumber","customerName","country"]])

right
````

## 5. `pd.merge()` — OUTER JOIN (FULL OUTER)

> **언제 쓰나?** 양쪽 테이블의 **모든 행을 남기고** 싶을 때
> 매칭이 안 되는 쪽은 `NaN` 으로 채워짐

```
INNER  ⊂  LEFT  ⊂  OUTER
INNER  ⊂  RIGHT ⊂  OUTER
```


````python
outer = pd.merge(
    orders[["orderNumber","customerNumber","status"]],
    customers[["customerNumber","customerName","country"]],
    on="customerNumber",
    how="outer",
)

print(f"OUTER JOIN 결과: {len(outer)}행")
print(f"  주문 없는 고객    : {outer['orderNumber'].isna().sum()}건")
print(f"  고객 정보 없는 주문: {outer['customerName'].isna().sum()}건")

outer
````

---
### JOIN 유형 비교 요약

| JOIN 유형 | 결과에 포함되는 행 | NaN 발생 |
|-----------|------------------|---------|
| `inner`   | 양쪽 모두 키가 있는 행만 | 없음 |
| `left`    | 왼쪽 전부 + 오른쪽 매칭 | 오른쪽 |
| `right`   | 오른쪽 전부 + 왼쪽 매칭 | 왼쪽 |
| `outer`   | 양쪽 전부 | 양쪽 |

## 6. 다중 키 & 체인 merge

### 6-1. 다중 키 merge
> 두 컬럼을 동시에 키로 사용 — `on=["col1", "col2"]`


````python
# 다중 키 merge: orderdetails + products
detail_product = pd.merge(
    orderdetails,
    products[["productCode","productName","productLine","buyPrice","MSRP"]],
    on="productCode",
    how="left",
)

detail_product["margin_pct"] = (
    (detail_product["priceEach"] - detail_product["buyPrice"])
    / detail_product["buyPrice"] * 100
).round(1)

detail_product[["orderNumber","productName","productLine","quantityOrdered","priceEach","margin_pct"]]
````

### 6-2. left_on / right_on — 컬럼명이 다를 때


````python
# employees self-join: reportsTo -> employeeNumber (매니저 정보)

emp_with_manager = pd.merge(
    employees,
    employees[["employeeNumber","firstName","lastName","jobTitle"]].rename(columns={
        "employeeNumber": "mgrNumber",
        "firstName":      "mgrFirstName",
        "lastName":       "mgrLastName",
        "jobTitle":       "mgrTitle",
    }),
    left_on="reportsTo",
    right_on="mgrNumber",
    how="left",
)

emp_with_manager[["firstName","lastName","jobTitle","mgrFirstName","mgrLastName","mgrTitle"]]
````

### 6-3. 체인 merge — 3개 이상 테이블 연결


````python
# orderdetails -> products -> productlines 체인
# 주문 상세에 제품 카테고리 설명까지 붙이기

chain = (
    orderdetails
    .merge(products[["productCode","productName","productLine"]], on="productCode")
    .merge(productlines[["productLine","textDescription"]],       on="productLine")
)

# 카테고리별 매출 합계
sales_by_line = (
    chain
    .assign(revenue=lambda d: d["quantityOrdered"] * d["priceEach"])
    .groupby("productLine", as_index=False)["revenue"]
    .sum()
    .sort_values("revenue", ascending=False)
)

sales_by_line
````

## 8. `pd.merge_ordered()` — 정렬 보장 병합

> **언제 쓰나?** 시계열/순서가 있는 데이터를 병합하되,
> **결과를 정렬된 순서로** 유지하고 싶을 때
> `fill_method='ffill'` 로 누락값을 앞 값으로 채우기 가능

```python
pd.merge_ordered(left, right, on=None, left_on=None, right_on=None,
                 fill_method=None, how='outer')
```


````python
# 시나리오: 주문 날짜와 결제 날짜를 시간 순으로 정렬 병합
# 고객별로 "언제 주문하고, 언제 결제했나" 를 시간 흐름으로 확인

ord_slim = orders[["orderNumber","orderDate","customerNumber"]].copy()
pay_slim = payments[["customerNumber","paymentDate","amount"]].copy()

ordered = pd.merge_ordered(
    ord_slim,
    pay_slim,
    left_on="orderDate",
    right_on="paymentDate",
    left_by="customerNumber",   # 고객별로 별도 정렬
    how="outer",
)

ordered.dropna(subset=["customerNumber"])[
    ["customerNumber","orderDate","orderNumber","paymentDate","amount"]
].head(12)
````

## 9. `pd.merge_asof()` — 근사값(asof) 병합

> **언제 쓰나?** 정확히 일치하는 키가 없을 때,
> **가장 가까운 이전(또는 이후) 값**으로 매칭
> 예) 주문일 이전에 가장 최근에 한 결제 찾기, 주가 데이터와 이벤트 연결

```python
pd.merge_asof(left, right, on=None, by=None,
              direction='backward')   # 'backward' | 'forward' | 'nearest'
```
> 주의: 두 DataFrame 모두 `on` 컬럼 기준으로 **정렬되어 있어야 함**


````python
# 시나리오: 각 주문에 대해, 주문일 이전 가장 최근 결제 찾기

ord_sorted = orders[["orderNumber","customerNumber","orderDate"]].sort_values("orderDate")
pay_sorted = payments[["customerNumber","paymentDate","amount"]].sort_values("paymentDate")

asof = pd.merge_asof(
    ord_sorted,
    pay_sorted,
    left_on="orderDate",
    right_on="paymentDate",
    by="customerNumber",      # 같은 고객 내에서만 매칭
    direction="backward",     # 주문일 이전 결제 (기본값)
)

print("각 주문의 직전 결제 정보:")
asof[["orderNumber","customerNumber","orderDate","paymentDate","amount"]]
````

````python
# direction='nearest' : 가장 가까운 날짜 (이전/이후 무관)

asof_nearest = pd.merge_asof(
    ord_sorted, pay_sorted,
    left_on="orderDate", right_on="paymentDate",
    by="customerNumber",
    direction="nearest",
)
asof_nearest[["orderNumber","orderDate","paymentDate","amount"]]
````

## 10. `combine_first()` — 결측값 보완

> **언제 쓰나?** DataFrame A 에 `NaN` 이 있을 때, DataFrame B 의 값으로 채우기
> A를 우선하고, A가 NaN 인 곳만 B에서 가져옴

```python
df_A.combine_first(df_B)
```


````python
# 시나리오: 두 소스에서 수집한 제품 가격 데이터 병합
# source_a : 일부 MSRP 누락
# source_b : 일부 buyPrice 누락 -> 서로 보완

source_a = products.set_index("productCode")[["productName","buyPrice","MSRP"]].copy()
source_b = source_a.copy()

# 의도적으로 누락값 생성
source_a.loc["S10_1678", "MSRP"]     = np.nan
source_a.loc["S12_1099", "buyPrice"] = np.nan
source_b.loc["S10_1949", "MSRP"]     = np.nan
source_b["buyPrice"] = source_b["buyPrice"] * 1.05  # b는 가격이 조금 다름

print("=== source_a (NaN 있음) ===")
display(source_a.head(4))

print("=== source_b (NaN 있음, 가격이 조금 다름) ===")
display(source_b.head(4))

filled = source_a.combine_first(source_b)
print("=== combine_first 결과 (a 우선, a의 NaN만 b로 채움) ===")
filled.head(4)
````

## 11. `compare()` — 두 DataFrame 차이 비교

> **언제 쓰나?** 두 버전 간 **어떤 값이 바뀌었는지** 빠르게 확인
> 변경된 셀만 표시 (나머지는 `NaN`)

```python
df1.compare(df2, result_names=("v1", "v2"), keep_shape=False)
```


````python
# 시나리오: 주문 상태 업데이트 전후 비교

orders_v1 = orders.set_index("orderNumber")[["status","customerNumber"]].copy()
orders_v2 = orders_v1.copy()

# 일부 주문 상태 변경
orders_v2.loc[10100, "status"] = "Processing"
orders_v2.loc[10102, "status"] = "On Hold"
orders_v2.loc[10105, "status"] = "Cancelled"

diff = orders_v1.compare(orders_v2, result_names=("이전","이후"))
print(f"변경된 주문: {len(diff)}건")
diff
````

````python
# keep_shape=True : 변경 없는 행도 포함 (NaN 으로 표시)
diff_full = orders_v1.compare(orders_v2, result_names=("이전","이후"), keep_shape=True)
diff_full
````

## 최종 요약 — 언제 어떤 함수를?

| 상황 | 함수 | 핵심 파라미터 |
|------|------|-------------|
| 같은 구조 데이터를 세로로 쌓기 | `pd.concat([a,b])` | `axis=0, ignore_index=True` |
| 같은 인덱스 데이터를 가로로 붙이기 | `pd.concat([a,b], axis=1)` | `join='inner'/'outer'` |
| SQL 스타일 조인 (컬럼 키) | `pd.merge(a, b, on=..., how=...)` | `how='inner/left/right/outer'` |
| 컬럼명이 다를 때 조인 | `pd.merge(a, b, left_on=..., right_on=...)` | — |
| 인덱스 기반 조인 | `a.join(b, on=..., how=...)` | `on=` 으로 컬럼->인덱스 매칭 |
| 정렬 보장 + 시계열 병합 | `pd.merge_ordered(a, b, left_by=...)` | `fill_method='ffill'` |
| 가장 가까운 키로 매칭 | `pd.merge_asof(a, b, direction=...)` | `'backward'/'forward'/'nearest'` |
| NaN 보완 (A 우선) | `a.combine_first(b)` | — |
| 두 버전 차이 확인 | `a.compare(b)` | `result_names=, keep_shape=` |
