# 퀴즈 2.

## 문제

$v_1, v_2, w \in R^n$, $\langle v_1, w \rangle = 3$, $\langle v_2, w \rangle = -2$. Find $\langle 2v_1 - 3v_2, w \rangle$.

---

## 개념 정의 (p.32)

**내적의 선형성 (분배법칙)**

- $\langle v+u, w \rangle = \langle v,w \rangle + \langle u,w \rangle$
- $\langle \lambda v, w \rangle = \lambda \langle v,w \rangle$

---

## 풀이

(벡터의 선형 결합에 대한 내적의 분배법칙)

$$\langle 2v_1 - 3v_2, w \rangle = 2\langle v_1, w \rangle - 3\langle v_2, w \rangle = 2 \cdot 3 - 3 \cdot (-2) = 12$$

---

## 정답

$$\boxed{12}$$

```python
# 임의의 벡터로 검증
# 벡터 정의
v1 = np.array([1, 0, 0])
v2 = np.array([0, 1, 0])
w = np.array([3, -2, 0])

# 조건 확인
print(f"<v1, w> = {np.dot(v1, w)}") # 3
print(f"<v2, w> = {np.dot(v2, w)}") # -2

# 문제 풀이 : <2v1 -3v2, w>
result = np.dot(2*v1 - 3*v2, w)
print(f"\n<2v1 -3v2, w> = {result}")

# 분배법칙으로 검증
result_dist = 2* np.dot(v1, w) - 3 * np.dot(v2, w)
print(f"\n2<v1, w> - 3<v2, w> = {result_dist}")
```

# 퀴즈 3.

## 문제

$$A = \begin{pmatrix} 3 & 1 \\ 2 & 1 \end{pmatrix} \begin{pmatrix} 1 & -1 \\ -2 & 3 \end{pmatrix}$$

---

## 개념 정의

**행렬의 곱셈**

$$A_{ij} = \sum_k B_{ik} C_{kj}$$

- 첫 번째 행렬의 $i$번째 행과 두 번째 행렬의 $j$번째 열의 내적을 계산

---

## 풀이

$$\begin{pmatrix} 3 & 1 \\ 2 & 1 \end{pmatrix} \begin{pmatrix} 1 & -1 \\ -2 & 3 \end{pmatrix} = \begin{pmatrix} (3\cdot1 + 1\cdot(-2)) & (3\cdot(-1) + 1\cdot3) \\ (2\cdot1 + 1\cdot(-2)) & (2\cdot(-1) + 1\cdot3) \end{pmatrix}$$

**첫 번째 행, 첫 번째 열:**
$$3\cdot1 + 1\cdot(-2) = 3 - 2 = 1$$

**첫 번째 행, 두 번째 열:**
$$3\cdot(-1) + 1\cdot3 = -3 + 3 = 0$$

**두 번째 행, 첫 번째 열:**
$$2\cdot1 + 1\cdot(-2) = 2 - 2 = 0$$

**두 번째 행, 두 번째 열:**
$$2\cdot(-1) + 1\cdot3 = -2 + 3 = 1$$

---

## 정답

$$\begin{pmatrix} 3 & 1 \\ 2 & 1 \end{pmatrix} \begin{pmatrix} 1 & -1 \\ -2 & 3 \end{pmatrix} = \boxed{\begin{pmatrix} 1 & 0 \\ 0 & 1 \end{pmatrix}}$$

```python
# 코드
import numpy as np

A = np.array([[3, 1], [2, 1]])
B = np.array([[1, -1], [-2, 3]])

# 행렬 곱셈
result = np.dot(A, B)
print(f"A = \n{result}")
```

# 퀴즈 5.

## 문제

$$\begin{cases} x + 2y - 4z = 2 \\ x + 3y - 6z = 1 \\ x + 2y - 3z = 4 \end{cases}$$

---

## 개념 정의

### Gauss 소거법 (강의자료 49~50페이지)
- 첨가행렬을 **상삼각행렬** 형태로 변환
- 이후 **후진 대입법**으로 해를 구함
- 아래쪽 행의 계수만 0으로 만들면 됨

### Gauss-Jordan 소거법 (강의자료 54~55페이지)
- 첨가행렬을 **RREF(단위행렬)** 형태로 변환
- 후진 대입 없이 **해를 직접 읽어냄**
- 위아래 모든 행의 계수를 0으로 만들어야 함

---

## 방법 1: Gauss 소거법

### 첨가행렬 구성

$$\left[\begin{array}{ccc|c} 1 & 2 & -4 & 2 \\ 1 & 3 & -6 & 1 \\ 1 & 2 & -3 & 4 \end{array}\right]$$

---

### 1단계: $R_2 \leftarrow R_2 - R_1$, $R_3 \leftarrow R_3 - R_1$
> x계수(1행)가 1이므로, 2·3행의 x를 없애기 위해 1행을 뺌

**R2 계산:**
$$R_2 - R_1 = (1-1,\ 3-2,\ -6+4,\ 1-2) = (0,\ 1,\ -2,\ -1)$$

**R3 계산:**
$$R_3 - R_1 = (1-1,\ 2-2,\ -3+4,\ 4-2) = (0,\ 0,\ 1,\ 2)$$

$$\left[\begin{array}{ccc|c} 1 & 2 & -4 & 2 \\ 0 & 1 & -2 & -1 \\ 0 & 0 & 1 & 2 \end{array}\right]$$

---

### 2단계: 후진 대입법으로 해 계산

**3행에서 z 직접 읽기:**
$$z = 2$$

**2행에 z = 2 대입:**
$$y - 2z = -1 \implies y - 2(2) = -1 \implies y = 3$$

**1행에 y = 3, z = 2 대입:**
$$x + 2(3) - 4(2) = 2 \implies x + 6 - 8 = 2 \implies x = 4$$

---

## 방법 2: Gauss-Jordan 소거법

### 1단계: $R_2 \leftarrow R_2 - R_1$, $R_3 \leftarrow R_3 - R_1$
> (Gauss 소거법과 동일)

$$\left[\begin{array}{ccc|c} 1 & 2 & -4 & 2 \\ 0 & 1 & -2 & -1 \\ 0 & 0 & 1 & 2 \end{array}\right]$$

---

### 2단계: $R_1 \leftarrow R_1 - 2R_2$
> 1행의 y계수가 **2**이므로, 0으로 만들기 위해 2배한 R2를 뺌

**R1 계산:**
$$R_1 - 2R_2 = (1-0,\ 2-2,\ -4+4,\ 2+2) = (1,\ 0,\ 0,\ 4)$$

$$\left[\begin{array}{ccc|c} 1 & 0 & 0 & 4 \\ 0 & 1 & -2 & -1 \\ 0 & 0 & 1 & 2 \end{array}\right]$$

---

### 3단계: $R_2 \leftarrow R_2 + 2R_3$
> 2행의 z계수가 **-2**이므로, 0으로 만들기 위해 2배한 R3를 더함

**R2 계산:**
$$R_2 + 2R_3 = (0+0,\ 1+0,\ -2+2,\ -1+4) = (0,\ 1,\ 0,\ 3)$$

$$\left[\begin{array}{ccc|c} 1 & 0 & 0 & 4 \\ 0 & 1 & 0 & 3 \\ 0 & 0 & 1 & 2 \end{array}\right]$$

> 좌변이 단위행렬 → 해를 직접 읽어냄

---

## 두 방법 비교

| 구분 | Gauss 소거법 | Gauss-Jordan 소거법 |
|------|-------------|-------------------|
| 목표 행렬 | 상삼각행렬 | 단위행렬(RREF) |
| 해 구하는 방법 | 후진 대입법 필요 | 직접 읽어냄 |
| 행 연산 횟수 | 적음 | 많음 |
| 참조 페이지 | 49~50페이지 | 54~55페이지 |

---

## 정답

$$\boxed{x = 4,\quad y = 3,\quad z = 2}$$

```python
import numpy as np

# 계수행렬 A 벡터
A = np.array([
    [1, 2, -4],
    [1, 3, -6],
    [1, 2, -3]
])

# 상수 b 벡터
b = np.array([2, 1, 4])

# 연립방정식 풀기 (Ax = b)
x = np.linalg.solve(A, b)
print(x)

# 검증
print(f"\n검증: A @ x = {A @ x}")
print(f"정답 b = {b}")
```

```python
# 코드
import numpy as np

# ============================================================
# 연립방정식
# x + 2y - 4z = 2
# x + 3y - 6z = 1
# x + 2y - 3z = 4
# ============================================================

A = np.array([
    [1, 2, -4],
    [1, 3, -6],
    [1, 2, -3]
], dtype=float)

b = np.array([2, 1, 4], dtype=float)

# ------------------------------------------------------------
# 방법 1: Gauss 소거법 (직접 구현)
# ------------------------------------------------------------
def gauss_elimination(A, b):
    n = len(b)
    # 첨가행렬 구성
    Aug = np.hstack([A.copy(), b.copy().reshape(-1, 1)])
    print("=" * 50)
    print("[ Gauss 소거법 ]")
    print("=" * 50)
    print(f"\n[초기] 첨가행렬:\n{Aug}\n")

    # 전진 소거 (상삼각행렬 만들기)
    for i in range(n):
        for j in range(i+1, n):
            factor = Aug[j, i] / Aug[i, i]
            Aug[j] = Aug[j] - factor * Aug[i]
            print(f"R{j+1} ← R{j+1} - {factor:.0f}×R{i+1}")
            print(f"{Aug}\n")

    print(f"[상삼각행렬 완성]:\n{Aug}\n")

    # 후진 대입법
    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        x[i] = (Aug[i, -1] - np.dot(Aug[i, i+1:n], x[i+1:n])) / Aug[i, i]
        print(f"x{i+1} = {x[i]:.0f}")

    return x

# ------------------------------------------------------------
# 방법 2: Gauss-Jordan 소거법 (직접 구현)
# ------------------------------------------------------------
def gauss_jordan(A, b):
    n = len(b)
    Aug = np.hstack([A.copy(), b.copy().reshape(-1, 1)])
    print("\n" + "=" * 50)
    print("[ Gauss-Jordan 소거법 ]")
    print("=" * 50)
    print(f"\n[초기] 첨가행렬:\n{Aug}\n")

    for i in range(n):
        # 피벗을 1로 정규화
        Aug[i] = Aug[i] / Aug[i, i]

        # 위아래 행 모두 소거
        for j in range(n):
            if j != i:
                factor = Aug[j, i]
                Aug[j] = Aug[j] - factor * Aug[i]
                print(f"R{j+1} ← R{j+1} - {factor:.0f}×R{i+1}")
                print(f"{Aug}\n")

    print(f"[단위행렬(RREF) 완성]:\n{Aug}\n")
    return Aug[:, -1]

# ------------------------------------------------------------
# 방법 3: numpy 내장 함수
# ------------------------------------------------------------
def numpy_solve(A, b):
    print("\n" + "=" * 50)
    print("[ numpy.linalg.solve ]")
    print("=" * 50)
    x = np.linalg.solve(A, b)
    return x

# ------------------------------------------------------------
# 실행 및 결과 출력
# ------------------------------------------------------------
x1 = gauss_elimination(A, b)
x2 = gauss_jordan(A, b)
x3 = numpy_solve(A, b)

print("\n" + "=" * 50)
print("[ 최종 결과 비교 ]")
print("=" * 50)
print(f"Gauss 소거법        : x={x1[0]:.0f}, y={x1[1]:.0f}, z={x1[2]:.0f}")
print(f"Gauss-Jordan 소거법 : x={x2[0]:.0f}, y={x2[1]:.0f}, z={x2[2]:.0f}")
print(f"numpy.linalg.solve  : x={x3[0]:.0f}, y={x3[1]:.0f}, z={x3[2]:.0f}")

# 검증
print("\n[ 검증: A @ x = b ]")
print(f"A @ x = {A @ x3}")
print(f"b     = {b}")
```

  + 광고 채널 A(x)와 광고 채널 B(y)의 최적 집행 횟수 찾기
    - 제약조건 1 (예산) : 3x + 1y = 9 (총 예산 9백만원 소진)
    - 제약조건 2 (시간) : 1x + 2y = 8 (총 가용 시간 80시간 소진)

## 수학적 원리
- 제약 조건 속 최적의 해답 찾기
- 행렬과 벡터를 활용하여 Ax = b 형태의 수식으로 표현
- 방정식을 풀기 위해 미지수의 계수(행렬 A)와 우변의 상수(벡터 b)를 하나로 합친 첨가 행렬(Augmented Matrix) 구성
- 미지수를 규칙적으로 줄여나가는 Gauss 소거법
- 대각선 성분만 남기는 사다리꼴 행렬로 만드는 Gauss-Jordan 소거법

```python
# 선형 방정식 행렬 세팅(Ax = b)
# 계수 행렬 A
A_matrix = np.array([
    [3, 1],
    [1, 2]
])

# 상수 벡터 b
b_vector = np.array([9, 8])

# 선형방정식 풀이 (np.linalg.solve 활용)
solution = np.linalg.solve(A_matrix, b_vector)
x_sol, y_sol = solution
print(x_sol, y_sol) # A 광고채널에서는 2회 집행, B 광고채널에는 3회 집행

# 현재 여러분들의 생각은 ??????????
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 비즈니스 문제 설정:
# 광고 채널 A(x)와 광고 채널 B(y)의 최적 집행 횟수 찾기
# 제약조건 1 (예산): 3x + 1y = 9 (총 예산 9백만 원 소진)
# 제약조건 2 (시간): 1x + 2y = 8 (총 가용 시간 80시간 소진)

# 1. 선형방정식 행렬 세팅 (Ax = b)
# 계수 행렬 A
A_matrix = np.array([
    [3, 1],
    [1, 2]
])
# 상수 벡터 b
b_vector = np.array([9, 8])

# 2. 선형방정식 풀이 (np.linalg.solve 활용)
# 역행렬 및 소거법의 원리가 내부적으로 적용되어 해(Solution)를 찾습니다.
solution = np.linalg.solve(A_matrix, b_vector)
x_sol, y_sol = solution
print(x_sol, y_sol)

# 3. 시각화 (FDV 원칙: 잉크-데이터 비율 최적화 및 직관적 주석)
x_vals = np.linspace(0, 5, 100)

# 식 정리를 통한 y값 도출
# 1) y = 9 - 3x

y_vals_budget = 9 - 3 * x_vals
# 2) y = (8 - x) / 2
y_vals_time = (8 - x_vals) / 2

fig, ax = plt.subplots(figsize=(7, 5))

# 두 제약 조건(방정식) 시각화
ax.plot(x_vals, y_vals_budget, label='Budget Constraint (3x + y = 9)', color='#C44E52', linewidth=2)
ax.plot(x_vals, y_vals_time, label='Time Constraint (x + 2y = 8)', color='#55A868', linewidth=2)

# 두 선이 만나는 교점(최적해) 표시
ax.scatter(x_sol, y_sol, color='#333333', s=120, zorder=5)

# Tufte 스타일의 명확한 데이터 라벨링
ax.annotate(f'Optimal Allocation\n(Channel A={x_sol:.1f}, Channel B={y_sol:.1f})',
            xy=(x_sol, y_sol),
            xytext=(x_sol + 0.3, y_sol + 1.5),
            fontsize=10,
            fontweight='bold',
            arrowprops=dict(facecolor='#333333', arrowstyle='->', lw=1.5))

ax.set_xlim(0, 4)
ax.set_ylim(0, 10)
ax.set_xlabel("Ad Channel A Executions (x)", fontweight='bold')
ax.set_ylabel("Ad Channel B Executions (y)", fontweight='bold')
ax.set_title("Finding Optimal Solution via Linear Equations", pad=15, fontweight='bold')

# 차트 정리
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.legend(frameon=False, loc='upper right')
ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

- 수학적인 의미
  + 두 선형 방정식 3x + y = 9, x + 2y = 8을 동시에 만족하는 유일한 해(solution)가 기하학적으로 두 직선의 교점과 같음
- 비즈니스적 의미
  + 한정된 예산과 시간을 최대한 활용하면서, 최대의 효율을 내기 위해, 광고 채널 A에 2회 집행, 채널 B에 3회 집행 => 최적 자원 배분

### 시나리오 변경
- 기존 제약 조건 : 예선 9, 시간 8
- 새로운 제약 조건 : 예산 12, 시간 7

==> 광고 채널 A에 몇 회 집행? 광고 채널 B 몇 회 집행?

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 선형방정식 행렬 세팅 (계수 행렬은 동일)
A_matrix = np.array([
    [3, 1],
    [1, 2]
])

# 2. 시나리오 1: 기존 제약 조건 (예산 9, 시간 8)
b_old = np.array([9, 8])
sol_old = np.linalg.solve(A_matrix, b_old)
x_old, y_old = sol_old

# 3. 시나리오 2: 새로운 제약 조건 (예산 12로 증가, 시간 7로 단축)
b_new = np.array([12, 7])
sol_new = np.linalg.solve(A_matrix, b_new)
x_new, y_new = sol_new

# 4. 시각화 (FDV 원칙: 전/후 비교를 위한 대비, 직접 라벨링)
x_vals = np.linspace(0, 6, 100)

# 방정식 정리 (y 도출)
y_budget_old = 9 - 3 * x_vals
y_time_old = (8 - x_vals) / 2

y_budget_new = 12 - 3 * x_vals
y_time_new = (7 - x_vals) / 2

fig, ax = plt.subplots(figsize=(8, 6))

# 과거 제약 조건 시각화 (흐린 색상 및 점선으로 처리하여 배경 정보로 활용)
ax.plot(x_vals, y_budget_old, color='#C44E52', linestyle='--', alpha=0.3)
ax.plot(x_vals, y_time_old, color='#55A868', linestyle='--', alpha=0.3)
ax.scatter(x_old, y_old, color='gray', s=80, alpha=0.5, zorder=3)

# 새로운 제약 조건 시각화 (선명한 색상과 굵은 선으로 강조)
ax.plot(x_vals, y_budget_new, color='#C44E52', linewidth=2)
ax.plot(x_vals, y_time_new, color='#55A868', linewidth=2)
ax.scatter(x_new, y_new, color='#333333', s=130, zorder=5)

# FDV 직접 표기법 (Direct Labeling) - 시선의 분산 방지
ax.text(1.0, 8.5, 'New Budget Constraint\n(3x + y = 12)', color='#C44E52', fontweight='bold', rotation=-67)
ax.text(3.5, 2.0, 'New Time Constraint\n(x + 2y = 7)', color='#55A868', fontweight='bold', rotation=-25)

# 최적해 이동 궤적 화살표 및 주석
ax.annotate('', xy=(x_new, y_new), xytext=(x_old, y_old),
            arrowprops=dict(facecolor='#333333', shrink=0.08, width=1.5, headwidth=8))

ax.annotate(f'Old Optimal\n({x_old:.1f}, {y_old:.1f})',
            xy=(x_old, y_old), xytext=(x_old - 1.2, y_old - 0.5), color='gray')

ax.annotate(f'New Optimal\n({x_new:.1f}, {y_new:.1f})',
            xy=(x_new, y_new), xytext=(x_new + 0.2, y_new + 0.3), fontweight='bold')

# 차트 정리
ax.set_xlim(0, 5)
ax.set_ylim(0, 10)
ax.set_xlabel("Ad Channel A Executions (x)", fontweight='bold')
ax.set_ylabel("Ad Channel B Executions (y)", fontweight='bold')
ax.set_title("Sensitivity Analysis: Shifting Constraints and New Optimal Solution", pad=20, fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(True, linestyle=':', alpha=0.5)

plt.tight_layout()
plt.show()
```
