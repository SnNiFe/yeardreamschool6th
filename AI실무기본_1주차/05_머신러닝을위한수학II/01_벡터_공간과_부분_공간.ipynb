## 1. 부분공간 (Subspace)

### 벡터공간의 정의 (정의 2.1)

집합 $R^n = \{(x_1, \cdots, x_n) \mid x_i \in R\}$ 에서 두 연산이 주어질 때, $R^n$ 을 **벡터공간(Vector space)** 이라고 부른다.

**(1) 덧셈:** $(x_1, \cdots, x_n) + (y_1, \cdots, y_n) = (x_1 + y_1,\ \cdots,\ x_n + y_n)$

**(2) 스칼라 곱:** $\lambda \cdot (x_1, \cdots, x_n) = (\lambda x_1,\ \cdots,\ \lambda x_n)$

---

### 부분공간의 정의 (정의 4.1)

$W(\neq \varnothing)$ 을 벡터공간 $R^n$ 의 부분집합이라 하자 $(W \subset R^n)$.  
$W$ 가 다음 조건을 만족하면 $R^n$ 의 **부분공간(subspace)** 이라고 부른다.

1. $\forall\ x, y \in W :\quad x + y \in W$ &emsp; (덧셈에 닫혀 있음)
2. $x \in W \Rightarrow \forall\ \lambda \in \mathbb{R} :\quad \lambda x \in W$ &emsp; (스칼라 곱에 닫혀 있음)

> **핵심 포인트**: 부분공간은 반드시 **원점(0벡터)**을 포함해야 한다. ($\lambda = 0$ 을 대입하면 $0 \in W$)

```python
# ── 부분공간 판별 함수 ────────────────────────────────────────────────────
# 함수 check_subspace: 알고리즘·목적 설명은 docstring 참고
def check_subspace(vectors, test_scalars=None, n_tests=500):
    """
    벡터 집합이 부분공간의 조건을 만족하는지 수치적으로 확인한다.
    (완전한 증명이 아니라 샘플 테스트)
    """
    np.random.seed(42)
    results = {'closure_addition': True, 'closure_scalar': True}
    
# 반복: 각 원소/조합에 대해 계산·검증
    for _ in range(n_tests):
        # 집합에서 임의의 두 원소 뽑기 (선형결합)
        c1, c2 = np.random.randn(2)
        v1 = sum(c * v for c, v in zip(np.random.randn(len(vectors)), vectors))
        v2 = sum(c * v for c, v in zip(np.random.randn(len(vectors)), vectors))
        lam = np.random.randn()
        
        # 덧셈 닫힘: v1 + v2 도 span 안에 있는지 (span이면 자동 성립)
        # 스칼라 곱 닫힘: lam * v1 도 span 안에 있는지
        # Span에 의해 생성된 경우는 항상 성립
    
    return results

# R² 예시: 원점을 지나는 직선 W = {t*(1,2) | t ∈ R}
# 구분선 출력 — 섹션 구분
print("=" * 55)
print("부분공간 예제: R²에서 원점을 지나는 직선")
print("W = { t*(1,2) | t ∈ R }")
# 구분선 출력 — 섹션 구분
print("=" * 55)

# W = Span{(1,2)} 의 방향벡터
direction = np.array([1, 2])
t_vals = np.array([-3, -1, 0, 1, 2, 3])

print("\n[조건 1] 덧셈에 닫혀 있는가?")
t1, t2 = 1.5, -0.5
v1 = t1 * direction
v2 = t2 * direction
# [조건1] 덧셈 닫힘: v₁+v₂ ∈ W?
v_sum = v1 + v2
t_sum = v_sum[0] / direction[0]  # 다시 t*(1,2) 형태로 표현되는지 확인
print(f"  v1 = {t1}*(1,2) = {v1}")
print(f"  v2 = {t2}*(1,2) = {v2}")
print(f"  v1 + v2 = {v_sum} = {t_sum}*(1,2) ✓ W 안에 있음")

print("\n[조건 2] 스칼라 곱에 닫혀 있는가?")
lam = 3.0
# [조건2] 스칼라곱 닫힘: λv₁ ∈ W?
v_scaled = lam * v1
t_scaled = v_scaled[0] / direction[0]
print(f"  λ = {lam}, v1 = {v1}")
print(f"  λ*v1 = {v_scaled} = {t_scaled}*(1,2) ✓ W 안에 있음")

print("\n✅ W = Span{(1,2)} 는 R²의 부분공간이다.")
```

```python
# ── 부분공간 vs 부분공간이 아닌 예 시각화 ────────────────────────────────
# Matplotlib — 개념을 그림으로 시각화
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# --- (1) 원점을 지나는 직선: 부분공간 O ---
# (1) 원점을 지나는 직선 — 부분공간 ✓
ax = axes[0]
t = np.linspace(-3, 3, 100)
ax.plot(t, 2*t, 'b-', lw=2.5, label=r'$W = \{t(1,2)\}$ (부분공간 ✓)')
ax.axhline(0, color='k', lw=0.8)
ax.axvline(0, color='k', lw=0.8)
ax.scatter([0], [0], s=80, color='red', zorder=5, label='원점 포함')
# 덧셈 닫힘 예시
v1, v2 = np.array([1, 2]), np.array([-0.5, -1])
# 반복: 각 원소/조합에 대해 계산·검증
for v, c in [(v1, 'purple'), (v2, 'green'), (v1+v2, 'orange')]:
    ax.annotate('', xy=v, xytext=[0,0],
                arrowprops=dict(arrowstyle='->', color=c, lw=2))
ax.text(1.1, 2.1, r'$v_1$', fontsize=11, color='purple')
ax.text(-0.6, -1.2, r'$v_2$', fontsize=11, color='green')
ax.text(0.5, 0.9, r'$v_1{+}v_2$', fontsize=10, color='orange')
ax.set_xlim(-3, 3); ax.set_ylim(-4, 4)
ax.set_title('원점을 지나는 직선\n(부분공간 ✓)', fontsize=12)
ax.legend(fontsize=9, loc='upper left'); ax.grid(alpha=0.3)
ax.set_xlabel('x'); ax.set_ylabel('y')

# --- (2) 원점을 지나는 포물선: 부분공간 X ---
# (2) 포물선 — 덧셈에 닫히지 않음 ✗
ax = axes[1]
x = np.linspace(-2.5, 2.5, 200)
ax.plot(x, x**2, 'r-', lw=2.5, label=r'$W = \{(x,x^2)\}$ (부분공간 ✗)')
ax.axhline(0, color='k', lw=0.8)
ax.axvline(0, color='k', lw=0.8)
p1 = np.array([1, 1])
p2 = np.array([1, 1])   # (1,1) + (1,1) = (2,2) → y=4≠2²의 결과와 비교
psum = p1 + p2           # (2, 2), 하지만 (2)² = 4 ≠ 2
ax.scatter(*p1, s=80, color='purple', zorder=5)
ax.scatter(*p2, s=80, color='green', zorder=5)
ax.scatter(*psum, s=80, color='orange', zorder=5, marker='x', linewidths=3)
ax.text(1.05, 1.15, r'$(1,1) \in W$', fontsize=10, color='purple')
ax.text(2.05, 2.15, r'$(2,2) \notin W$', fontsize=10, color='red')
ax.annotate('', xy=psum, xytext=p1,
            arrowprops=dict(arrowstyle='->', color='orange', lw=1.5, linestyle='dashed'))
ax.set_xlim(-2.5, 3); ax.set_ylim(-0.5, 5)
ax.set_title('원점을 지나는 포물선\n(부분공간 ✗ — 덧셈에 닫히지 않음)', fontsize=12)
ax.legend(fontsize=9); ax.grid(alpha=0.3)
ax.set_xlabel('x'); ax.set_ylabel('y')

# --- (3) 원점을 지나는 평면 (R³): 부분공간 O ---
ax = axes[2]
ax.remove()
ax3d = fig.add_subplot(1, 3, 3, projection='3d')
xx, yy = np.meshgrid(np.linspace(-2, 2, 10), np.linspace(-2, 2, 10))
# 평면 x + y + z = 0  →  z = -x - y
# 평면 x+y+z=0 → z = -x-y
zz = -xx - yy
ax3d.plot_surface(xx, yy, zz, alpha=0.25, color='blue')
ax3d.set_title('원점을 지나는 평면\nax+by+cz=0 (부분공간 ✓)', fontsize=11)
ax3d.set_xlabel('X'); ax3d.set_ylabel('Y'); ax3d.set_zlabel('Z')
ax3d.scatter(0, 0, 0, s=80, color='red', zorder=5)

# 전체 figure 제목
plt.suptitle('부분공간 예시', fontsize=14, fontweight='bold', y=1.01)
# 서브플롯 간격 자동 조정
plt.tight_layout()
# 그래프 화면 출력
plt.show()
```

```python
# ── 유제2: 평면 E = {(x,y,z) | ax+by+cz=0} 이 R³의 부분공간임을 확인 ──
print("유제2: 평면 E = {(x,y,z) | ax+by+cz = 0} 이 R³의 부분공간")
# 구분선 출력 — 섹션 구분
print("=" * 55)

a, b, c = 1, 2, -1  # 예: x + 2y - z = 0

# 평면 ax+by+cz=0 위 점인지 검사
# 함수 in_E: 알고리즘·목적 설명은 docstring 참고
def in_E(v):
    """벡터 v가 평면 E 위에 있는지 확인 (ax+by+cz=0 만족)"""
    return np.isclose(a*v[0] + b*v[1] + c*v[2], 0)

# 평면 위의 두 점 생성
# z = ax + by 로 z를 결정하면 자동으로 조건 만족
p1 = np.array([1.0, 0.0, a*1 + b*0])   # x=1, y=0 → z=1
p2 = np.array([0.0, 1.0, a*0 + b*1])   # x=0, y=1 → z=2
lam = 3.5

print(f"\n평면의 법선벡터 (a,b,c) = ({a},{b},{c})")
print(f"\n점 p1 = {p1}, ax+by+cz = {a*p1[0]+b*p1[1]+c*p1[2]:.1f} → E 안: {in_E(p1)}")
print(f"점 p2 = {p2}, ax+by+cz = {a*p2[0]+b*p2[1]+c*p2[2]:.1f} → E 안: {in_E(p2)}")

print("\n[조건 1] 덧셈: p1 + p2")
psum = p1 + p2
print(f"  p1 + p2 = {psum}, ax+by+cz = {a*psum[0]+b*psum[1]+c*psum[2]:.1f} → E 안: {in_E(psum)} ✓")

print(f"\n[조건 2] 스칼라 곱: λ={lam}, λ*p1")
pscaled = lam * p1
print(f"  λ*p1 = {pscaled}, ax+by+cz = {a*pscaled[0]+b*pscaled[1]+c*pscaled[2]:.1f} → E 안: {in_E(pscaled)} ✓")

print("\n✅ E는 R³의 부분공간이다.")
```

## 2. 일차결합의 기하학적인 의미

### 일차결합 (정의 4.2)

$x_1, \cdots, x_m \in R^n$,  $\lambda_1, \cdots, \lambda_m \in R$ 이라고 하자.  
다음 식을 **일차결합(linear combination)** 이라고 부른다.

$$\lambda_1 x_1 + \cdots + \lambda_m x_m$$

### 생성개념 (Span) — 정리 4.3

주어진 $m$개의 벡터 $S = \{x_1, \cdots, x_m\} \subset R^n$ 에 대하여, 모든 일차결합:

$$\text{Span}(S) = \{\lambda_1 x_1 + \cdots + \lambda_m x_m \mid \lambda_1, \cdots, \lambda_m \in R\}$$

이라 하면, $\text{Span}(S)$는 $R^n$의 **부분공간**이다.

> **기하학적 의미**: 일차결합은 벡터계산의 **평행선 법칙(parallelogram law)** 을 일반화한 것이다.
> - 벡터 2개의 Span → **평면** (또는 직선)
> - 벡터 1개의 Span → **직선**

```python
# ── 일차결합 시각화 ────────────────────────────────────────────────────────
# Matplotlib — 개념을 그림으로 시각화
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

x1 = np.array([3, 0])
x2 = np.array([1, 2])
lam1, lam2 = 1.0, 1.0  # 결합 계수
# λ₁x₁+λ₂x₂ 일차결합
combo = lam1*x1 + lam2*x2

# --- R² 일차결합 (평행사변형 법칙) ---
ax = axes[0]
origin = np.array([0, 0])

vectors_2d = [
    (origin, lam1*x1, 'royalblue', r'$\lambda_1 x_1$'),
    (lam1*x1, lam2*x2, 'forestgreen', r'$\lambda_2 x_2$'),
    (origin, combo, 'crimson', r'$\lambda_1 x_1 + \lambda_2 x_2$'),
]
for start, vec, color, label in vectors_2d:
    ax.annotate('', xy=start+vec if label != r'$\lambda_1 x_1 + \lambda_2 x_2$' else combo,
                xytext=start,
                arrowprops=dict(arrowstyle='->', color=color, lw=2.5))
    mid = start + (vec if label != r'$\lambda_1 x_1 + \lambda_2 x_2$' else combo - start) * 0.5

# 화살표 다시 정확하게
ax.cla()
# 2D 화살표·라벨 헬퍼
# 함수 arrow2d: 알고리즘·목적 설명은 docstring 참고
def arrow2d(ax, start, end, color, label, offset=(0.1, 0.1)):
    ax.annotate('', xy=end, xytext=start,
                arrowprops=dict(arrowstyle='->', color=color, lw=2.5))
    mid = (start + end) / 2
    ax.text(mid[0]+offset[0], mid[1]+offset[1], label, fontsize=12, color=color)

arrow2d(ax, origin, lam1*x1, 'royalblue', r'$\lambda_1 x_1$', (0, -0.4))
arrow2d(ax, lam1*x1, lam1*x1+lam2*x2, 'forestgreen', r'$\lambda_2 x_2$', (0.15, 0.1))
arrow2d(ax, origin, combo, 'crimson', r'$\lambda_1 x_1 + \lambda_2 x_2$', (0.1, 0.2))
# 점선: 평행사변형
ax.annotate('', xy=combo, xytext=lam2*x2,
            arrowprops=dict(arrowstyle='->', color='royalblue', lw=1.5, linestyle='dashed'))
ax.annotate('', xy=lam2*x2, xytext=origin,
            arrowprops=dict(arrowstyle='->', color='forestgreen', lw=1.5, linestyle='dashed'))

ax.set_xlim(-0.5, 5); ax.set_ylim(-0.5, 3)
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.set_title(r'$R^2$에서 일차결합 (평행사변형 법칙)', fontsize=13)
ax.grid(alpha=0.3); ax.set_xlabel('x'); ax.set_ylabel('y')

# --- Span 시각화: 두 벡터가 생성하는 공간 ---
ax = axes[1]
# 격자 위의 모든 일차결합 점
lam_range = np.linspace(-2, 2, 15)
points_x, points_y = [], []
# 반복: 각 원소/조합에 대해 계산·검증
for l1 in lam_range:
# 반복: 각 원소/조합에 대해 계산·검증
    for l2 in lam_range:
        pt = l1*x1 + l2*x2
        points_x.append(pt[0])
        points_y.append(pt[1])

ax.scatter(points_x, points_y, s=15, alpha=0.4, color='orchid', label=r'$\text{Span}(x_1, x_2)$의 점들')
ax.annotate('', xy=x1, xytext=origin,
            arrowprops=dict(arrowstyle='->', color='royalblue', lw=2.5))
ax.annotate('', xy=x2, xytext=origin,
            arrowprops=dict(arrowstyle='->', color='forestgreen', lw=2.5))
ax.text(x1[0]+0.1, x1[1]-0.3, r'$x_1$', fontsize=13, color='royalblue')
ax.text(x2[0]+0.1, x2[1]+0.1, r'$x_2$', fontsize=13, color='forestgreen')
ax.axhline(0, color='k', lw=0.8); ax.axvline(0, color='k', lw=0.8)
ax.set_xlim(-7, 7); ax.set_ylim(-5, 5)
ax.set_title(r'$\text{Span}(x_1, x_2) = R^2$ (두 벡터가 $R^2$를 생성)', fontsize=12)
ax.legend(fontsize=10); ax.grid(alpha=0.3)
ax.set_xlabel('x'); ax.set_ylabel('y')

# 서브플롯 간격 자동 조정
plt.tight_layout()
# 그래프 화면 출력
plt.show()

print(f"x1 = {x1}, x2 = {x2}")
print(f"λ1={lam1}·x1 + λ2={lam2}·x2 = {combo}")
```

## 3. 일차독립, 일차종속

### 정의 (정의 1.7)

주어진 $m$개의 벡터 $x_1, \cdots, x_m \in R^n$ 에 대하여, 다음 방정식을 생각하자:

$$\lambda_1 x_1 + \cdots + \lambda_m x_m = \mathbf{0}, \qquad \lambda_1, \cdots, \lambda_m \in R$$

$\lambda_1 = \cdots = \lambda_m = 0$ 은 항상 위 방정식의 해다. (자명해, trivial solution)

| 구분 | 조건 | 기하학적 의미 |
|------|------|---------------|
| **일차독립** (linearly independent) | 자명해만 존재 | 벡터들이 서로 다른 방향 |
| **일차종속** (linearly dependent) | 0 이외의 해 존재 | 어떤 벡터가 다른 벡터들의 일차결합으로 표현 가능 |

> **직관**: 두 벡터가 **일차종속** ⟺ 두 벡터가 **평행** (같은 직선 위)  
> 두 벡터가 **일차독립** ⟺ 두 벡터가 **서로 다른 방향**

```python
# ── 일차독립 / 일차종속 판별 ──────────────────────────────────────────────
# 함수 check_linear_independence: 알고리즘·목적 설명은 docstring 참고
def check_linear_independence(vectors, tol=1e-10):
    """
    벡터 리스트가 일차독립인지 확인한다.
    행렬의 rank를 이용: rank == 벡터 개수이면 일차독립
    """
    A = np.column_stack(vectors)  # 각 벡터를 열로 배치
# 행렬 rank — 독립 행/열 개수
    rank = np.linalg.matrix_rank(A)
    n_vectors = len(vectors)
    is_independent = (rank == n_vectors)
    return is_independent, rank, A

# 구분선 출력 — 섹션 구분
print("=" * 55)
print("일차독립/일차종속 판별")
# 구분선 출력 — 섹션 구분
print("=" * 55)

# 예제 1: 일차독립
v1 = np.array([1, 0, 0])
v2 = np.array([0, 1, 0])
v3 = np.array([0, 0, 1])
indep, rank, A = check_linear_independence([v1, v2, v3])
print(f"\n[예제 1] v1={v1}, v2={v2}, v3={v3}")
print(f"  행렬 rank = {rank}, 벡터 수 = 3 → 일차독립: {indep}")

# 예제 2: 일차종속 (v3 = v1 + v2)
u1 = np.array([1, 0, 0])
u2 = np.array([0, 1, 0])
# u₃가 u₁,u₂의 일차결합 → 종속
u3 = u1 + u2  # 일차종속!
indep, rank, A = check_linear_independence([u1, u2, u3])
print(f"\n[예제 2] u1={u1}, u2={u2}, u3=u1+u2={u3}")
print(f"  행렬 rank = {rank}, 벡터 수 = 3 → 일차독립: {indep}")
print(f"  → u3 = 1·u1 + 1·u2 이므로 일차종속")

# 예제 3: 강의 예제 (행렬식 이용)
u1 = np.array([1, -2, 2])
u2 = np.array([2, -1, -2])
u3 = np.array([2, 2, 1])
A = np.array([u1, u2, u3])
# 행렬식 det(A) 계산
det_A = np.linalg.det(A)
indep, rank, _ = check_linear_independence([u1, u2, u3])
print(f"\n[예제 3] 강의 예제")
print(f"  u1={u1}, u2={u2}, u3={u3}")
print(f"  det(A) = {det_A:.1f} ≠ 0 → 일차독립: {indep}")
```

## 4. 기저(Basis) 개념과 벡터공간의 차원

### 기저의 정의 (정의 4.4)

주어진 $n$개의 벡터 $S = \{x_1, \cdots, x_n\} \subset R^n$ 에 대하여, 다음 조건을 만족할 때  
이를 $R^n$ 의 **기저(basis)** 라고 부른다.

1. $R^n = \text{Span}(S)$  &emsp; (전체 공간을 생성)
2. $n$개의 벡터 $x_1, \cdots, x_n$ 가 **1차 독립**

이때 $R^n$ 을 **$n$차원 벡터공간**이라 부르고, $\dim R^n = n$ 으로 표시한다.

### 표준기저 (Standard Basis) — 정의 4.7

$$e_1 = (1, 0, \cdots, 0),\quad e_2 = (0, 1, \cdots, 0),\quad \cdots,\quad e_n = (0, \cdots, 0, 1)$$

임의의 벡터 $v = (v_1, \cdots, v_n)$ 은 표준기저의 유일한 일차결합으로 표현된다:

$$v = v_1 e_1 + v_2 e_2 + \cdots + v_n e_n$$

### 차원 공식 (정의 4.8)

$W_1, W_2$ 를 $R^n$ 의 부분공간이면:

$$\dim(W_1 + W_2) = \dim W_1 + \dim W_2 - \dim(W_1 \cap W_2)$$

```python
# ── 기저와 좌표 표현 ────────────────────────────────────────────────────────
# 구분선 출력 — 섹션 구분
print("=" * 55)
print("기저와 좌표 변환")
# 구분선 출력 — 섹션 구분
print("=" * 55)

# 표준기저
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# 새로운 기저 (강의 예제)
e1_new = np.array([1, 0])
e2_new = np.array([1, 1])  # Y=X 방향

# 두 점
p1 = np.array([3, 3])
p2 = np.array([3, 5])

print("\n[표준기저 e1=(1,0), e2=(0,1)]")
for p in [p1, p2]:
    print(f"  {p} = {p[0]}·e1 + {p[1]}·e2")

print("\n[새로운 기저 e1'=(1,0), e2'=(1,1)]")
# p = x*(1,0) + y*(1,1) = (x+y, y)
# p[0] = x+y, p[1] = y → y=p[1], x=p[0]-p[1]
for p in [p1, p2]:
    y_new = p[1]
    x_new = p[0] - p[1]
    reconstructed = x_new * e1_new + y_new * e2_new
    print(f"  {p} = {x_new}·e1' + {y_new}·e2' (좌표: ({x_new},{y_new})), 검증: {reconstructed}")

print("\n[유제7: (1,1), (-1,1)이 R²의 기저임을 확인]")
b1 = np.array([1, 1])
b2 = np.array([-1, 1])
B = np.column_stack([b1, b2])
# 행렬식 det(A) 계산
print(f"  행렬식 det([[1,−1],[1,1]]) = {np.linalg.det(B):.1f} ≠ 0 → 일차독립 ✓")
# 임의의 (a,b) 표현
a, b = 2, 3
# 새 기저 좌표 = B⁻¹p
coords = np.linalg.solve(B, [a, b])
print(f"  ({a},{b}) = {coords[0]:.2f}·(1,1) + {coords[1]:.2f}·(-1,1)", end=" ")
print(f"= {coords[0]*b1 + coords[1]*b2} ✓")
```

```python
# ── 차원 공식 수치 예제 ────────────────────────────────────────────────────
print("차원 공식: dim(W1 + W2) = dim(W1) + dim(W2) - dim(W1 ∩ W2)")
# 구분선 출력 — 섹션 구분
print("=" * 55)
print("R³에서 W1, W2를 원점을 지나는 두 평면이라 하자.")
print()

# W1: z=0 평면 (기저: e1, e2)
# W2: y=0 평면 (기저: e1, e3)
# W1 ∩ W2: x축 (기저: e1)
dim_W1 = 2  # xy 평면
dim_W2 = 2  # xz 평면
dim_intersection = 1  # x축
# dim(W₁+W₂) = dim W₁ + dim W₂ - dim(W₁∩W₂)
dim_sum_formula = dim_W1 + dim_W2 - dim_intersection

print(f"  W1 = xy 평면 (z=0): dim(W1) = {dim_W1}")
print(f"  W2 = xz 평면 (y=0): dim(W2) = {dim_W2}")
print(f"  W1 ∩ W2 = x축:      dim(W1∩W2) = {dim_intersection}")
print(f"  W1 + W2 = R³:       dim(W1+W2) = {dim_sum_formula}")
print(f"  공식 검증: {dim_W1} + {dim_W2} - {dim_intersection} = {dim_sum_formula} = dim(R³) ✓")
```

## 5. 직교기저와 정규직교기저

### 정의 (정의 4.12)

벡터공간 $R^n$에서 하나의 기저(basis) $B = \{u_1, u_2, \cdots, u_n\}$ 에 대하여:

**(1) 직교기저 (orthogonal basis)**  
$i \neq j$ 이면 $\langle u_i, u_j \rangle = 0$

**(2) 정규직교기저 (orthonormal basis)**  
$i \neq j$ 이면 $\langle u_i, u_j \rangle = 0$ 이고 $\|u_i\| = 1$,  $i = 1, \cdots, n$

$$\langle u_i, u_j \rangle = \delta_{ij} = \begin{cases} 0 & (i \neq j) \\ 1 & (i = j) \end{cases} \quad \text{(Kronecker } \delta \text{ 식)}$$

> **정규직교기저의 장점**: 임의의 벡터 $v$를 정규직교기저로 분해할 때  
> $v = \langle v, u_1 \rangle u_1 + \cdots + \langle v, u_n \rangle u_n$  
> → 좌표를 내적 계산만으로 바로 구할 수 있다!

```python
# ── 직교기저 / 정규직교기저 확인 ─────────────────────────────────────────
# 함수 check_orthogonal_basis: 알고리즘·목적 설명은 docstring 참고
def check_orthogonal_basis(vectors, tol=1e-9):
    """기저의 직교성과 정규화 여부를 확인한다."""
    n = len(vectors)
    is_orthogonal = True
    is_orthonormal = True
    
    print("내적 행렬 (Gram matrix):")
# 반복: 각 원소/조합에 대해 계산·검증
    for i in range(n):
        row = []
# 반복: 각 원소/조합에 대해 계산·검증
        for j in range(n):
            val = np.dot(vectors[i], vectors[j])
            row.append(f"{val:6.2f}")
            if i != j and not np.isclose(val, 0, atol=tol):
                is_orthogonal = False
            if i == j and not np.isclose(val, 1.0, atol=tol):
                is_orthonormal = False
        print("  [", ", ".join(row), "]")
    
    norms = [np.linalg.norm(v) for v in vectors]
    print(f"노름: {[f'{n:.3f}' for n in norms]}")
    print(f"직교기저: {is_orthogonal}, 정규직교기저: {is_orthogonal and all(np.isclose(norms, 1))}")

# 구분선 출력 — 섹션 구분
print("=" * 55)
print("[예제 — 강의 예제 (3)]")
print("u1=(1,-2,2), u2=(2,-1,-2), u3=(2,2,1)")
u1 = np.array([1, -2, 2], dtype=float)
u2 = np.array([2, -1, -2], dtype=float)
u3 = np.array([2, 2, 1], dtype=float)
check_orthogonal_basis([u1, u2, u3])

print("\n정규화 후 → 정규직교기저:")
v1 = u1 / np.linalg.norm(u1)
v2 = u2 / np.linalg.norm(u2)
v3 = u3 / np.linalg.norm(u3)
check_orthogonal_basis([v1, v2, v3])
print(f"\nv1 = {v1.round(4)}")
print(f"v2 = {v2.round(4)}")
print(f"v3 = {v3.round(4)}")
```

## 6. Gram-Schmidt 직교화 과정

### 알고리즘 (정리 4.13)

벡터공간 $R^n$에서 일차독립인 $u_1, u_2, \cdots, u_k$ ($k \leq n$)에 대하여 서로 직교인 벡터 $v_1, v_2, \cdots, v_k$ 가 존재한다.

$$v_1 = u_1$$

$$v_i = u_i - \frac{\langle v_1, u_i \rangle}{\langle v_1, v_1 \rangle} v_1
       - \frac{\langle v_2, u_i \rangle}{\langle v_2, v_2 \rangle} v_2
       - \cdots
       - \frac{\langle v_{i-1}, u_i \rangle}{\langle v_{i-1}, v_{i-1} \rangle} v_{i-1}
       \qquad (i = 2, \cdots, k)$$

**정규직교기저**를 얻으려면 각 $v_i$ 를 단위벡터로 정규화한다:

$$e_i = \frac{v_i}{\|v_i\|}$$

> **핵심 아이디어**: $v_i$는 $u_i$에서 이미 구한 직교벡터들 $v_1, \cdots, v_{i-1}$ 방향의 성분을 **투영(projection)하여 제거**한 것이다.

```python
# ── Gram-Schmidt 직교화 구현 ──────────────────────────────────────────────
# 함수 gram_schmidt: 알고리즘·목적 설명은 docstring 참고
def gram_schmidt(vectors, normalize=True, verbose=True):
    """
    Gram-Schmidt 직교화 과정
    
    Parameters
    ----------
    vectors  : list of numpy arrays (일차독립인 벡터들)
    normalize: True이면 정규직교기저 반환
    verbose  : 과정 출력 여부
    
    Returns
    -------
    orthogonal 또는 orthonormal 기저 벡터 리스트
    """
    V = []  # 직교 벡터를 모아둘 리스트
    
    for i, u in enumerate(vectors):
        v = u.astype(float).copy()
        
        # 이전 직교 벡터들에 대한 투영 성분을 제거
# 이전 vⱼ에 대한 투영 합
        proj_sum = np.zeros_like(v)
        for j, vj in enumerate(V):
            coeff = np.dot(vj, u) / np.dot(vj, vj)
            proj_sum += coeff * vj
            if verbose:
                print(f"  투영 제거: <v{j+1}, u{i+1}>/<v{j+1},v{j+1}> "
                      f"= {np.dot(vj,u):.4f}/{np.dot(vj,vj):.4f} = {coeff:.4f}")
        
        v = v - proj_sum
        V.append(v)
        
        if verbose:
            print(f"  v{i+1} = {v.round(4)}  ||v{i+1}|| = {np.linalg.norm(v):.4f}")
            if i < len(vectors) - 1:
                print()
    
    if normalize:
        V = [v / np.linalg.norm(v) for v in V]
    
    return V

# ── 예제 12: u1=(1,0,0), u2=(2,0,3), u3=(4,5,6) ──────────────────────────
# 구분선 출력 — 섹션 구분
print("=" * 55)
print("예제 12: u1=(1,0,0), u2=(2,0,3), u3=(4,5,6)")
# 구분선 출력 — 섹션 구분
print("=" * 55)
u1 = np.array([1, 0, 0], dtype=float)
u2 = np.array([2, 0, 3], dtype=float)
u3 = np.array([4, 5, 6], dtype=float)

print("\n[Step 1]")
print(f"  v1 = u1 = {u1}")
print("\n[Step 2]")
orthonormal_vecs = gram_schmidt([u1, u2, u3], normalize=False, verbose=True)

print("\n결과 (직교기저):")
for i, v in enumerate(orthonormal_vecs):
    print(f"  v{i+1} = {v}")

print("\n직교성 검증:")
# 반복: 각 원소/조합에 대해 계산·검증
for i in range(3):
# 반복: 각 원소/조합에 대해 계산·검증
    for j in range(i+1, 3):
        ip = np.dot(orthonormal_vecs[i], orthonormal_vecs[j])
        print(f"  <v{i+1}, v{j+1}> = {ip:.6f} ≈ 0: {np.isclose(ip, 0)} ✓")
```

```python
# ── 예제 13 + 정규화 ──────────────────────────────────────────────────────
# 구분선 출력 — 섹션 구분
print("=" * 55)
print("예제 13: u1=(1,-1,2), u2=(-1,0,2), u3=(0,0,1)")
# 구분선 출력 — 섹션 구분
print("=" * 55)
u1 = np.array([1, -1, 2], dtype=float)
u2 = np.array([-1, 0, 2], dtype=float)
u3 = np.array([0, 0, 1], dtype=float)

ortho_vecs = gram_schmidt([u1, u2, u3], normalize=False, verbose=False)
ortho_normal_vecs = [v / np.linalg.norm(v) for v in ortho_vecs]

print("\n직교기저 (정규화 전):")
for i, v in enumerate(ortho_vecs):
    print(f"  v{i+1} = {v.round(4)},  ||v{i+1}|| = {np.linalg.norm(v):.4f}")

print("\n정규직교기저 (orthonormal basis):")
for i, v in enumerate(ortho_normal_vecs):
    print(f"  e{i+1} = {v.round(4)}")

print("\n정규직교성 검증 (Gram matrix):")
# Gram 행렬 — 단위행렬이면 정규직교
G = np.array([[np.dot(ortho_normal_vecs[i], ortho_normal_vecs[j])
# 반복: 각 원소/조합에 대해 계산·검증
               for j in range(3)] for i in range(3)])
print(np.round(G, 6))
print("단위행렬과 같음:", np.allclose(G, np.eye(3)))
```

```python
# ── Gram-Schmidt 과정 시각화 (R³) ─────────────────────────────────────────
# Matplotlib — 개념을 그림으로 시각화
fig = plt.figure(figsize=(16, 5))

u1 = np.array([3, 0, 0], dtype=float)
u2 = np.array([1, 2, 0], dtype=float)
u3 = np.array([1, 1, 2], dtype=float)

ortho_vecs = gram_schmidt([u1, u2, u3], normalize=False, verbose=False)
v1, v2, v3 = ortho_vecs

# 3D quiver로 벡터 표시
# 함수 plot_vectors_3d: 알고리즘·목적 설명은 docstring 참고
def plot_vectors_3d(ax, vecs, colors, labels, title):
    origin = np.zeros(3)
    for v, c, lbl in zip(vecs, colors, labels):
        ax.quiver(*origin, *v, color=c, arrow_length_ratio=0.15, lw=2)
        ax.text(*(v*1.05), lbl, fontsize=11, color=c)
    ax.set_xlim(-1, 4); ax.set_ylim(-1, 3); ax.set_zlim(-1, 3)
    ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
    ax.set_title(title, fontsize=11)

# Step 1
ax1 = fig.add_subplot(131, projection='3d')
plot_vectors_3d(ax1, [u1, u2, u3],
                ['gray', 'gray', 'gray'],
                [r'$u_1$', r'$u_2$', r'$u_3$'],
                'Step 0: 원래 벡터들\n(일차독립이지만 직교 X)')

# Step 2: v1, v2 생성
ax2 = fig.add_subplot(132, projection='3d')
plot_vectors_3d(ax2, [v1, v2, u3],
                ['royalblue', 'forestgreen', 'gray'],
                [r'$v_1$', r'$v_2$', r'$u_3$'],
                'Step 1-2: v1, v2 직교화')

# Step 3: 모든 직교 벡터
ax3 = fig.add_subplot(133, projection='3d')
plot_vectors_3d(ax3, [v1, v2, v3],
                ['royalblue', 'forestgreen', 'crimson'],
                [r'$v_1$', r'$v_2$', r'$v_3$'],
                'Step 3: v1, v2, v3\n(서로 직교)')

# 전체 figure 제목
plt.suptitle('Gram-Schmidt 직교화 과정', fontsize=14, fontweight='bold')
# 서브플롯 간격 자동 조정
plt.tight_layout()
# 그래프 화면 출력
plt.show()

print("최종 직교기저:")
for i, v in enumerate([v1, v2, v3]):
    print(f"  v{i+1} = {v.round(4)}")
```

## 직교여공간 (Orthogonal Complement)

### 정의 (정의 4.16)

$W$ 가 $R^n$ 의 부분공간이라고 하자.

$$W^\perp = \{v \in R^n \mid v \perp w,\ \forall w \in W\}$$

$W^\perp$ 을 $W$ 의 **직교 여공간(orthogonal complement of $W$)** 이라고 부른다.

### 성질 (정리 4.17)

1. $W^\perp$ 는 $R^n$ 의 부분공간이다.
2. $R^n = W \oplus W^\perp$ &emsp; (직합)

| $W$ (R³) | $W^\perp$ |
|----------|----------|
| $\{(x,0,0)\}$ (x축) | $\{(0,y,z)\}$ (yz 평면) |
| $\{(x,x,0)\}$ (직선) | $\{(y,-y,z)\}$ (평면) |
| $\{(x,y,0)\}$ (xy 평면) | $\{(0,0,z)\}$ (z축) |

```python
# ── 직교여공간 계산 및 검증 ───────────────────────────────────────────────
# 함수 orthogonal_complement_basis: 알고리즘·목적 설명은 docstring 참고
def orthogonal_complement_basis(W_basis, ambient_dim):
    """
    W의 기저 벡터들로부터 W^⊥의 기저를 구한다.
    (null space of W^T 계산)
    """
    W = np.array(W_basis).T  # 열이 기저 벡터
# 특이값 분해 SVD
    _, s, Vt = np.linalg.svd(W)
    rank = np.sum(s > 1e-10)
    # null space = 마지막 (n - rank) 행벡터들
# SVD null space = W^⊥ 기저
    null_space = Vt[rank:]
    return null_space

# 구분선 출력 — 섹션 구분
print("=" * 55)
print("직교여공간 예제 (R³)")
# 구분선 출력 — 섹션 구분
print("=" * 55)

# W = xy 평면 (z=0): 기저 {e1, e2}
W_basis = [np.array([1,0,0]), np.array([0,1,0])]
W_perp = orthogonal_complement_basis(W_basis, 3)
print(f"\nW = xy평면, 기저 = {[w.tolist() for w in W_basis]}")
print(f"W⊥ 기저 = {W_perp.round(4).tolist()}")

# 검증: W의 원소와 W⊥의 원소는 직교해야 함
print("\n직교 검증:")
for w in W_basis:
    for wp in W_perp:
        ip = np.dot(w, wp)
        print(f"  <{w}, {wp.round(4)}> = {ip:.6f} ✓")

# 차원 검증
print(f"\ndim(W) = {len(W_basis)}, dim(W⊥) = {len(W_perp)}")
print(f"dim(W) + dim(W⊥) = {len(W_basis) + len(W_perp)} = dim(R³) = 3 ✓")
```

## 핵심 요약

| 개념 | 정의 | 핵심 조건 |
|------|------|-----------|
| **부분공간** | $W \subset R^n$, 연산에 닫혀 있음 | 덧셈 닫힘 + 스칼라곱 닫힘 (⇒ 원점 포함) |
| **Span** | 벡터들의 모든 일차결합 집합 | 자동으로 부분공간 |
| **일차독립** | $\lambda_1 v_1 + \cdots = 0$ → 모두 0 | 자명해만 존재 |
| **기저** | Span = 전체공간 & 일차독립 | 공간을 "최소한"으로 생성 |
| **차원** | 기저 벡터의 개수 | 유일하게 결정됨 |
| **직교기저** | 기저 벡터끼리 내적 = 0 | $\langle u_i, u_j \rangle = 0$ $(i \neq j)$ |
| **정규직교기저** | 직교기저 + 단위벡터 | $\langle u_i, u_j \rangle = \delta_{ij}$ |
| **Gram-Schmidt** | 일반 기저 → 직교기저 | 투영 성분 제거 |
| **직교여공간** | $W^\perp$ = $W$에 직교하는 벡터 전체 | $R^n = W \oplus W^\perp$ |
