## 수식으로 문제 풀이

### 문제 1. 총매출 함수 R(p)와 총이윤 함수 π(p) 도출

**총매출 R(p):**

가격 × 판매량이므로

$$R(p) = p \cdot Q(p) = p(1000 - 50p) = 1000p - 50p^2$$

**총비용을 가격 p의 함수로 변환:**

$$C(Q(p)) = 5Q(p) + 2000 = 5(1000-50p) + 2000 = 5000 - 250p + 2000 = 7000 - 250p$$

**총이윤 π(p):**

$$\pi(p) = R(p) - C(Q(p)) = (1000p - 50p^2) - (7000 - 250p)$$

$$\boxed{\pi(p) = -50p^2 + 1250p - 7000}$$

이윤 함수는 p에 대한 **위로 볼록한(concave) 2차함수**입니다.

---

### 문제 2. 1차 최적화 조건(FOC) 및 도함수

**1차 최적화 조건(First-Order Condition)이란:**

함수가 극값(최댓값 또는 최솟값)을 가지려면, 그 지점에서 함수의 **접선의 기울기(도함수)가 0**이어야 합니다. 이윤 함수 π(p)가 위로 볼록(2차항 계수가 음수)하므로, 도함수가 0이 되는 지점은 곧 **이윤이 최대가 되는 지점**입니다. 즉,

$$\frac{d\pi}{dp} = 0$$

을 만족하는 p가 이윤 극대화의 후보점(critical point)이 됩니다.

**도함수 계산:**

$$\pi(p) = -50p^2 + 1250p - 7000$$

각 항을 미분하면 ($\frac{d}{dp}(-50p^2) = -100p$, $\frac{d}{dp}(1250p) = 1250$, 상수항은 0):

$$\boxed{\frac{d\pi}{dp} = -100p + 1250}$$

(참고: 2차 도함수 $\frac{d^2\pi}{dp^2} = -100 < 0$ 이므로 이 점은 최댓값임이 보장됩니다 — 2차 조건(SOC) 충족)

---

### 문제 3. 최적 가격 p*와 최적 판매량 Q*

**최적 가격 p\* 계산:** FOC에 따라 도함수를 0으로 설정

$$-100p + 1250 = 0$$

$$100p = 1250$$

$$\boxed{p^* = 12.5}$$

**최적 판매량 Q\* 계산:** 수요함수에 p* 대입

$$Q^* = 1000 - 50(12.5) = 1000 - 625$$

$$\boxed{Q^* = 375}$$

---

### 문제 4. 최대 순이윤 π*

이윤 함수에 p* = 12.5 대입:

$$\pi^* = -50(12.5)^2 + 1250(12.5) - 7000$$

$$\pi^* = -50(156.25) + 15625 - 7000$$

$$\pi^* = -7812.5 + 15625 - 7000$$

$$\boxed{\pi^* = 812.5}$$
