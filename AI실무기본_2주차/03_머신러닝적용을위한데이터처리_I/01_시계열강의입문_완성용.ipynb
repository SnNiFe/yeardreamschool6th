## 시계열 분해 및 시각화
- 시계열 분석의 꽃인 STL(Seasonal-Trend decomposition using LOESS) 기법을 사용하여 데이터를 4가지 성분으로 완벽하게 분리하고 시각화하는 과정입니다.
- STL 분해 결과 해석 가이드
  + Trend (추세): 데이터의 본질적인 방향성을 보여줍니다. 주식이나 실적 데이터라면 우상향하는지, 혹은 특정 시점부터 꺾이는지 파악할 수 있습니다.
  + Seasonal (계절성): 원본 데이터에서 파동처럼 보이던 부분이 분리된 것입니다. 값이 일정하다면 고정된 패턴이 있는 것이고, 진폭이 변한다면 계절적 영향력이 바뀌고 있음을 의미합니다.
  + Residuals (잔차): 이 그래프에서 갑자기 튀는 값(Spike)이 있다면, 그것은 단순한 계절성이나 추세로 설명할 수 없는 특이 사건(Event)이 발생했음을 시사합니다.

```python
# 1. STL 분해 실행
# period=4: 데이터의 계절성 주기 설정 (예: 분기별 데이터라면 4, 월별 데이터라면 12)
# .fit(): 모델을 데이터에 맞추어 성분(Trend, Seasonal, Resid)을 계산함
advanced_decomposition = STL(df.data, period=4).fit()

# 2. 4행 1열 구조의 서브플롯 생성
# sharex=True: 모든 그래프가 동일한 x축(날짜)을 공유하도록 설정하여 비교를 용이하게 함
fig, (ax1, ax2, ax3, ax4) = plt.subplots(nrows=4, ncols=1, sharex=True, figsize=(10, 8))

# 3. 각 성분별 시각화
# [Observed]: 원본 데이터 (추세 + 계절성 + 잔차의 합)
ax1.plot(advanced_decomposition.observed)
ax1.set_ylabel('Observed')

# [Trend]: 장기적인 상승 또는 하락 경향 (계절성과 노이즈가 제거된 매끄러운 곡선)
ax2.plot(advanced_decomposition.trend)
ax2.set_ylabel('Trend')

# [Seasonal]: 특정 주기마다 반복되는 패턴 (여기서는 period=4에 따른 분기 패턴)
ax3.plot(advanced_decomposition.seasonal)
ax3.set_ylabel('Seasonal')

# [Residuals]: 추세와 계절성으로 설명되지 않는 무작위 변동 (이상치 탐지에 활용)
ax4.plot(advanced_decomposition.resid)
ax4.set_ylabel('Residuals')

# 4. x축 눈금 및 라벨 설정 (이전 시각화와 동일하게 연도 매핑)
plt.xticks(np.arange(0, 85, 8), [1960, 1962, 1964, 1966, 1968, 1970, 1972, 1974, 1976, 1978, 1980])

# 5. 가독성 개선 및 레이아웃 조정
fig.autofmt_xdate() # 날짜 라벨 겹침 방지
plt.tight_layout() # 서브플롯 간 간격 자동 조정
plt.show()
```

## 평가지표
- MAPE 함수 만들기
- 단점 (Zero Division): 실제값(y_true) 중에 0이 포함되어 있으면 계산 결과가 무한대(inf)가 되거나 에러가 발생합니다. (이럴 때는 분모에 아주 작은 값을 더하거나 SMAPE 같은 대안 지표를 사용합니다.)

```python
def mape(y_true, y_pred):
    """
    모델의 예측 오차를 백분율(%)로 반환하는 함수

    Args:
        y_true: 실제값 (Numpy array 또는 Pandas Series)
        y_pred: 모델의 예측값 (Numpy array 또는 Pandas Series)
    """
    # 1. (y_true - y_pred) / y_true: 실제값 대비 오차의 비율 계산
    # 2. np.abs(...): 오차가 양수든 음수든 크기만을 고려하기 위해 절대값 취함
    # 3. np.mean(...): 모든 데이터 포인트에 대한 오차 비율의 평균 계산
    # 4. * 100: 소수점 형태의 오차율을 백분율(%) 단위로 변환
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100
```

# 확률보행 따라가보기
- "어제와 오늘의 연결" (자기상관)
  + 확률보행에서 오늘의 값($y_t$)은 어제의 값($y_{t-1}$)에 무작위 오차($\epsilon_t$)를 더해서 결정됩니다.
$$y_t = y_{t-1} + \epsilon_t$$
  + 즉, 현재 위치는 바로 이전 위치에서 시작하며, 어디로 튈지는 오직 '확률'에 맡깁니다.
- 예측 불가능성 (Efficiency)
  + 다음 단계가 오를지 내릴지는 순전히 무작위 오차($\epsilon_t$)에 달려 있습니다.
  + 이 오차는 보통 평균이 0인 정규분포를 따른다고 가정
  + "내일의 가장 좋은 예측치는 오늘이다"라는 결론에 도달하게 됩니다.
- 비정상성 (Non-stationary)
  + 확률보행 데이터는 시간이 지날수록 평균에서 멀어지거나 변동폭이 커지는 경향이 있습니다.
  + 즉, 일정한 평균값으로 회귀하려는 성질이 없기 때문에 통계적으로 '불안정한' 데이터입니다.

## 확률보행 시각화

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 재현성을 위한 랜덤 시드 고정
# 동일한 시드 번호(42)를 사용하면 실행할 때마다 같은 무작위 결과가 나옵니다.
np.random.seed(42)

# 2. 무작위 단계(Steps) 생성
# np.random.standard_normal(1000): 평균 0, 표준편차 1인 표준정규분포에서 1000개의 난수 생성
steps = np.random.standard_normal(1000)

# 시작점 설정: 첫 번째 위치를 0으로 고정하여 원점에서 시작하게 함
steps[0] = 0

# 3. 확률보행 데이터 생성 (누적 합계)
# np.cumsum(steps): 이전 단계의 모든 값을 더해 나가는 방식
# 오늘 값 = 어제 값 + 오늘 발생한 무작위 오차(step)
random_walk = np.cumsum(steps)

# 4. 시각화 설정
fig, ax = plt.subplots()

# 생성된 확률보행 데이터를 선 그래프로 출력
ax.plot(random_walk)

# 축 레이블 설정
ax.set_xlabel('Timesteps')  # 시간의 흐름 (단계)
ax.set_ylabel('Value')      # 해당 시점의 누적 가치

# 그래프 여백 자동 최적화 및 출력
plt.tight_layout()
plt.show()
```

## 정상성 테스트

### 정상 vs 비정상 시계열 시각화
- $\alpha = 0.5$ (정상성)
  + 값이 일시적으로 튀더라도 다음 단계에서 절반($0.5$)으로 줄어듭니다.
  + 따라서 시간이 지나도 일정한 평균(0 근처) 주위에서 머무는 평균 회귀(Mean Reversion) 특성을 보입니다.
- $\alpha = 1.0$ (비정상성)
  + 이전의 모든 충격(노이즈)이 그대로 다음 단계로 전달됩니다.
  + 이것이 바로 확률보행(Random Walk)이며, 데이터가 정해진 평균 없이 정처 없이 떠돌게 됩니다.

```python
import numpy as np

def simulate_process(is_stationary: bool) -> np.array:
    """
    정상성(Stationary) 또는 비정상성(Non-stationary) 시계열 데이터를 생성하는 시뮬레이션 함수입니다.

    수학적 모델 (1차 자기회귀 모형, AR(1)):
    y_t = alpha * y_{t-1} + e_t  (단, e_t ~ N(0, 1)인 백색 잡음)

    Args:
        is_stationary (bool):
            - True: 정상성 시계열 (자기회귀계수 |alpha| < 1) 생성
            - False: 비정상성 시계열 (단위근을 갖는 확률보행, alpha = 1) 생성

    Returns:
        np.array: 생성된 400개의 시계열 데이터 포인트를 담은 1차원 배열
    """

    # 난수 생성기의 시드를 고정하여, 실험 시마다 동일한 확률적 패턴(시계열 궤적)이 재현되도록 보장합니다.
    np.random.seed(42)

    # 길이가 400인 초기화되지 않은 NumPy 배열 생성 (성능 최적화를 위한 메모리 사전 할당)
    process = np.empty(400)

    if is_stationary:
        # [1] 정상성 시계열 (Stationary Process) 설정
        # 자기회귀계수(alpha)를 0.5로 설정합니다 (|alpha| < 1).
        # 정상성 시계열은 '평균 회귀(Mean-reversion)' 성질을 가지며,
        # 과거에 발생한 외부 충격(Shock)이 시간이 지남에 따라 점차 소멸하는 특징이 있습니다.
        alpha = 0.5
        process[0] = 0       # 시계열의 초기 싯점(y_0) 값을 0으로 설정

    else:
        # [2] 비정상성 시계열 (Non-stationary Process / Random Walk) 설정
        # 자기회귀계수(alpha)를 1로 설정합니다 (이를 단위근, Unit Root 과정이라 함).
        # 확률보행(Random Walk) 과정에서는 과거의 백색 잡음(오차)이 영구적으로 누적됩니다.
        # 따라서 분산이 시간에 비례하여 무한히 증가하며, 특정 평균값으로 회귀하지 않습니다.
        alpha = 1
        process[0] = 10      # 시계열의 초기 싯점(y_0) 값을 10으로 임의 설정

    # [3] 시계열 데이터 생성 루프
    # AR(1) 점화식인 y_{t} = alpha * y_{t-1} + noise 를 순차적으로 적용합니다.
    for i in range(400):
        # 배열의 인덱스 초과(IndexError)를 방지하기 위한 조건문
        if i + 1 < 400:
            # 다음 시점(i+1)의 값을 계산:
            # (현재 시점의 값 * alpha) + (평균 0, 표준편차 1인 표준정규분포 백색 잡음 추가)
            process[i+1] = alpha * process[i] + np.random.standard_normal()
        else:
            # 배열의 마지막 인덱스에 도달하여 데이터 포인트 400개가 모두 채워지면 루프 종료
            break

    return process
```

```python
stationary = simulate_process(True)
non_stationary = simulate_process(False)
```

```python
# 1. 그래프 도화지(Figure)와 축(Axis) 객체 생성
fig, ax = plt.subplots()

# 2. 정상성(Stationary) 데이터 플로팅
# linestyle='-': 실선으로 표시하여 평균으로 회귀하는 안정적인 흐름을 강조
ax.plot(stationary, linestyle='-', label='stationary')

# 3. 비정상성(Non-stationary) 데이터 플로팅
# linestyle='--': 점선으로 표시하여 정처 없이 떠도는(Random Walk) 특성을 시각적으로 구분
ax.plot(non_stationary, linestyle='--', label='non-stationary')

# 4. 축 레이블 설정
ax.set_xlabel('Timesteps')  # 시간의 흐름 (단계)
ax.set_ylabel('Value')      # 관측값

# 5. 범례(Legend) 표시
# loc=2: 범례를 '좌측 상단(Upper Left)'에 배치하여 데이터 선과 겹침을 최소화
ax.legend(loc=2)

# 6. 레이아웃 최적화 및 그래프 출력
# tight_layout(): 축 레이블이나 제목이 잘리지 않도록 여백을 자동으로 조정
plt.tight_layout()
plt.show()
```

### 평균이 일정한가?

- 시계열 분석에서 매우 중요한 개념인 정상성(Stationarity)의 첫 번째 조건, 즉 '시간에 따라 평균이 일정한가?'를 시각적으로 검증하는 훌륭한 방법입니다.

- **1. 에르고딕성(Ergodicity)과 평균의 수렴**
정상성 시계열에서 시간 평균은 앙상블 평균(Ensemble Average)과 같습니다.
- 즉, 시간이 무한대로 갈 때 누적 평균은 이론적 기댓값에 수렴합니다.
$$\lim_{T \to \infty} \frac{1}{T} \sum_{t=1}^{T} y_t = E[y_t] = \mu$$



```python
import numpy as np
import matplotlib.pyplot as plt

def mean_over_time(process: np.array) -> np.array:
    """
    시간의 흐름에 따른 시계열 데이터의 누적 평균(Cumulative Mean)을 계산하는 함수

    통계적 의미:
    정상성 시계열은 '에르고딕성(Ergodicity)'을 가지므로, 표본의 크기(시간 t)가
    커질수록 누적 평균이 모평균(True Mean)으로 수렴해야 합니다.
    """
    mean_func = []

    for i in range(len(process)):
        # 주의: i=0일 때 process[:0]은 빈 배열이 되어 np.mean() 계산 시
        # RuntimeWarning(Mean of empty slice)이 발생하고 NaN을 반환합니다.
        # (그래프를 그릴 때 Matplotlib이 NaN을 무시하므로 에러가 나진 않습니다.)
        mean_func.append(np.mean(process[:i]))

    return mean_func

# 정상성 / 비정상성 시계열 데이터의 누적 평균 계산
# (이전 simulate_process 함수로 생성된 결과물이 있다고 가정)
stationary_mean = mean_over_time(stationary)
non_stationary_mean = mean_over_time(non_stationary)

# ----------------- 시각화(Plotting) 부분 -----------------
# 그래프를 통해 두 시계열의 누적 평균이 시간에 따라 어떻게 달라지는지 비교합니다.

fig, ax = plt.subplots()

# 1. 정상성(Stationary) 시계열의 누적 평균 시각화
# 시간이 지남에 따라 초기 변동성이 줄어들고, 특정 값(이전 코드 기준 0)으로 수렴하는 모습을 보입니다.
ax.plot(stationary_mean, label='stationary')

# 2. 비정상성(Non-stationary) 시계열의 누적 평균 시각화
# 확률보행(Random Walk) 과정이므로 모평균으로 수렴하지 않고 계속해서 발산하거나 무작위로 표류(Drift)합니다.
ax.plot(non_stationary_mean, linestyle='--', label='non-stationary')

# 축 이름 및 범례 설정
ax.set_xlabel('Timesteps')  # x축: 시간의 흐름 (t)
ax.set_ylabel('Mean')       # y축: 해당 시점까지의 누적 평균
ax.legend(loc=1)            # 범례 위치 (1 = 우측 상단)

plt.tight_layout()          # 그래프 요소들이 겹치지 않도록 레이아웃 자동 조정
plt.show()                  # 그래프 출력
```

### 분산이 일정한가?

- 이 코드 역시 시계열 데이터가 정상성을 띠는지 직관적으로 판단하는 데 매우 유용합니다.

- **1. AR(1)과 Random Walk의 이론적 분산 차이**
작성하신 그래프의 결과를 수학적으로 풀어보면 다음과 같습니다. (여기서 $\sigma^2$은 백색 잡음의 분산인 1을 의미합니다.)

  + **정상성 시계열 (AR(1), $\alpha=0.5$)**: 이론적 분산은 $Var(y_t) = \frac{\sigma^2}{1 - \alpha^2}$ 입니다. 따라서 시간이 충분히 지나면 그래프의 실선은 $\frac{1}{1 - 0.5^2} = \frac{1}{0.75} \approx 1.33$ 부근으로 수렴하게 됩니다.
  + **비정상성 시계열 (Random Walk, $\alpha=1$):** 단위근 과정에서 분산은 $Var(y_t) = t\sigma^2$ 가 됩니다.
- 분산이 시간 $t$에 선형적으로 비례하므로 그래프의 점선은 시간이 갈수록 계속해서 우상향(발산)하게 됩니다.



```python
import numpy as np
import matplotlib.pyplot as plt

def var_over_time(process: np.array) -> np.array:
    """
    시간의 흐름에 따른 시계열 데이터의 누적 표본분산(Cumulative Variance)을 계산하는 함수

    통계적 의미:
    정상성 시계열은 '동분산성(Homoscedasticity)'을 가정합니다. 즉, 시간이 지나 데이터가 누적되어도
    표본분산이 발산하지 않고 이론적인 모분산(True Variance)에 수렴해야 합니다.
    """
    var_func = []

    for i in range(len(process)):
        # 주의: i=0일 때 process[:0]은 길이가 0인 배열이 되어 np.var() 계산 시
        # RuntimeWarning(Degrees of freedom <= 0 for slice)이 발생하고 NaN을 반환합니다.
        # i=1일 때는 데이터가 1개뿐이므로 분산은 0.0이 됩니다.
        var_func.append(np.var(process[:i]))

    return var_func

# 정상성 / 비정상성 시계열 데이터의 누적 분산 계산
stationary_var = var_over_time(stationary)
non_stationary_var = var_over_time(non_stationary)

# ----------------- 시각화(Plotting) 부분 -----------------
fig, ax = plt.subplots()

# 1. 정상성(Stationary) 시계열의 누적 분산 시각화
# 초기에는 변동이 있을 수 있으나, 시간이 지남에 따라 일정한 상숫값에 수렴하는 모습을 보입니다.
ax.plot(stationary_var, label='stationary')

# 2. 비정상성(Non-stationary) 시계열의 누적 분산 시각화
# 확률보행(Random Walk) 과정의 특징으로 인해 분산이 시간에 비례하여 끝없이 증가(발산)합니다.
ax.plot(non_stationary_var, linestyle='--', label='non-stationary')

# 축 이름 및 범례 설정
ax.set_xlabel('Timesteps')  # x축: 시간의 흐름 (t)
ax.set_ylabel('Variance')   # y축: 해당 시점까지의 누적 분산
ax.legend(loc=2)            # 범례 위치 (2 = 좌측 상단, 분산 그래프가 우상향하므로 좌측이 적합함)

plt.tight_layout()          # 여백 최적화
plt.show()                  # 그래프 출력
```

### 확률보행 확인

- 시계열 분석에서 정상성을 진단할 때 가장 널리 쓰이는 ADF 검정(Augmented Dickey-Fuller Test)을 수행하는 표준적인 방법입니다.

- ADF 검정을 완벽하게 이해하기 위해서는 가설의 구조를 명확히 아는 것이 중요합니다.

  + **귀무가설 ($H_0$):** 시계열 데이터에 단위근이 존재한다 (즉, 데이터는 **비정상성**이다).
  + **대립가설 ($H_1$):** 시계열 데이터에 단위근이 존재하지 않는다 (즉, 데이터는 **정상성**이다).

- **결과 해석 시나리오 (Random Walk 데이터 기준):**
입력하신 `random_walk` 데이터는 단위근 모형($\alpha=1$)이므로, 위 코드를 실행하면 다음과 같은 결과를 얻게 될 것입니다.

* **p-value가 0.05보다 크게 나옴:** (예: 0.45, 0.82 등)
* **해석:** 유의수준 **5%** 하에서 귀무가설($H_0$)을 기각할 충분한 통계적 근거가 없습니다. 따라서 해당 시계열은 **비정상성(Non-stationary)** 데이터라는 결론을 내리게 됩니다.

#### 자기상관함수(ACF, Autocorrelation Function) 시각화

- 현재 데이터가 과거의 데이터와 얼마나 강한 패턴을 공유하고 있는지(자기상관성)를 직관적으로 보여줍니다.
- 그래프를 실행하시면 위와 같이 가로축은 시차(Lag), 세로축은 상관계수(-1.0 ~ 1.0)를 나타내는 막대 그래프(Correlogram)가 출력될 것입니다.
- 이 그래프에서 주목해야 할 세 가지 핵심 포인트는 다음과 같습니다.
  이해하기 쉽도록 핵심 내용만 추려 단일 수준의 불릿 포인트(Bullet point)로 간결하게 정리해 드립니다.
  + **비정상성의 느린 감소(Slow Decay):** $y_t = y_{t-1} + \epsilon_t$ 구조로 인해 과거의 기억이 오래 유지되므로, 시차가 커져도 ACF 막대가 천천히 감소합니다. (반면, 정상성 데이터는 급격히 0으로 떨어집니다.)
  + **시차 0 (Lag 0)의 의미:** 현재 시점($y_t$) 자기 자신과의 상관관계를 의미하므로 첫 번째 막대는 항상 **1.0**입니다.
  + **95% 신뢰구간 (파란색 음영):** 막대가 이 영역을 **벗어나면** '통계적으로 유의미한 상관관계가 있음'을, 영역 **안에 있으면** '상관관계가 없음'을 뜻합니다. (랜덤 워크는 시차 20에서도 영역을 벗어나는 경우가 많습니다.)

```python
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt  # (참고) 이전 코드에서 plt를 import 했다고 가정합니다.

# 자기상관함수(ACF, Autocorrelation Function) 시각화
# ACF는 시계열의 현재 값과 일정 시차(Lag)만큼 떨어진 과거 값 사이의 선형 상관관계를 나타냅니다.
#
# random_walk: 앞서 생성하거나 검증한 비정상성 시계열 데이터
# lags=20: 현재 시점(t)으로부터 과거 20번째 시차(t-20)까지의 상관계수를 계산합니다.
# 주의: 코드 끝의 세미콜론(;)은 Jupyter Notebook 환경에서
# 텍스트 출력(AxesSubplot 객체 정보)을 숨기고 그래프만 깔끔하게 보여주는 실무적인 팁입니다.
plot_acf(random_walk, lags=20);

# 그래프 요소들이 겹치지 않도록 레이아웃을 자동 조정하고 화면에 출력합니다.
plt.tight_layout()
plt.show()
```

## 차분진행 후 시각화

- 비정상성 시계열이 가진 문제(일정하지 않은 평균과 발산하는 분산)를 해결하기 위해, 시계열 분석의 가장 핵심적인 전처리 기법인 차분(Differencing)을 적용
- **차분의 원리:** 수식 $y_t - y_{t-1} = \epsilon_t$ 를 적용하여, 비정상성 데이터를 평균과 분산이 일정한 순수 백색 잡음(White Noise)으로 변환합니다.
- **정상성 시각화:** 차분 후의 그래프는 더 이상 발산하거나 표류하지 않고, 0을 중심으로 일정하게 진동하는 정상성(Stationary) 패턴을 보입니다.
- **적분 차수(Order of Integration):** 1번 차분하여 정상성을 띠는 원본 데이터를 **I(1)** 과정, 차분이 완료되어 정상성을 획득한 상태를 **I(0)** 과정이라 하며, 이는 ARIMA 모델 적용을 위한 필수 조건입니다.


```python
import numpy as np
import matplotlib.pyplot as plt

# np.diff를 사용하여 시계열 데이터의 1차 차분(First Difference)을 계산합니다.
# 차분은 현재 시점의 값에서 바로 이전 시점의 값을 빼는 작업입니다 (y_t - y_{t-1}).
# n=1은 1차 차분을 의미하며, 결과 배열의 길이는 원본 데이터보다 1개 짧아집니다 (400 -> 399).
diff_random_walk = np.diff(random_walk, n=1)

# ----------------- 시각화(Plotting) 부분 -----------------
# 1차 차분된 데이터의 시계열 그래프를 그립니다.
plt.plot(diff_random_walk)

# 그래프 제목 설정: 차분된 랜덤 워크가 백색 잡음(White Noise) 형태로 변환됨을 보여줍니다.
plt.title('Differenced Random Walk')
plt.xlabel('Timesteps')
plt.ylabel('Value')

plt.tight_layout()
plt.show()
```

```python
ADF_result = adfuller(diff_random_walk)

print(f'ADF Statistic: {ADF_result[0]}')
print(f'p-value: {ADF_result[1]}')
```

#### 그래프 해석 : 백색 잡음(White Noise)

- **X축과 Y축의 의미:**
  + **X축 (0~20):** 시차(Lag)를 의미합니다. '0'은 현재 시점, '1'은 1스텝 과거, '20'은 20스텝 과거의 데이터를 뜻합니다.
  + **Y축 (-1.0 ~ 1.0):** 피어슨 상관계수(Correlation Coefficient)입니다. 1에 가까울수록 강한 양의 상관관계, -1에 가까울수록 강한 음의 상관관계, 0이면 상관관계가 없음을 나타냅니다.


- **시차 0 (Lag = 0)의 솟아오른 막대:**
  + 값이 정확히 **1.0**에 위치해 있습니다. 이는 현재 데이터($y_t$)와 동일한 현재 데이터($y_t$) 자기 자신과의 상관관계를 계산한 것이므로, 항상 1이 나오는 것이 정상입니다.


- **파란색 음영 영역 (신뢰구간):**
  + 이 구간은 95% 신뢰구간(Confidence Interval)입니다. 통계적으로 "이 구간 안에 막대가 위치하면, 해당 시차의 상관계수는 사실상 0과 다를 바 없다(유의미한 상관관계가 없다)"라고 해석하는 기준선이 됩니다.


- **시차 1~20 (Lag 1 ~ 20)의 막대 분포:**
  + 시차 1부터 20까지의 **모든 막대가 파란색 음영 영역 안**에 들어와 있으며, 0을 중심으로 아주 미세하게만 무작위로 흔들리고 있습니다.
  + 이는 1스텝 전의 과거 데이터든, 20스텝 전의 과거 데이터든 현재 데이터에 **아무런 영향을 주지 못하고 있음**을 의미합니다.

### 백색잡음의 비즈니스의 의미
- ARIMA(0, 1, 0) 즉 랜덤 워크 모델이 비즈니스 현장에서 가지는 의미

#### 1. 과거의 평균은 미래를 속인다 (단순 예측의 법칙)

- 정상성 데이터라면 평균으로 회귀하는 성질이 있으므로 과거의 '평균값'을 미래 예측에 쓰는 것이 타당할 수 있습니다.
- 하지만 확률보행(비정상성) 데이터는 평균이 시간에 따라 계속 떠돌아다닙니다(Drift).
- 따라서 과거 전체의 평균을 사용하면 예측이 완전히 빗나가게 됩니다.
- 통계적으로 증명된 가장 합리적인 미래 예측값은 평균이 아니라 바로 '오늘의 값(가장 최근 데이터)'입니다.
  + 이를 수식으로는 $\hat{y}_{t+1} = y_t$ 로 표현하며, 비즈니스에서는 이를 단순 예측(Naive Forecasting)이라고 부릅니다.
- **예시:** 내일의 삼성전자 주가를 예측하는 가장 좋은 기준점은 지난 10년 치 주가의 평균이 아니라, **오늘의 종가**입니다.

### 2. 효율적 시장 가설 (Efficient Market Hypothesis)

- 주가, 환율, 원자재 가격 같은 금융/경제 데이터라면 이러한 결과는 자주 나타남
- 시장에 참여하는 수많은 사람에 의해 모든 정보가 실시간으로 가격에 반영되는 시장에서는 과거의 가격 패턴(차트)만으로 미래를 예측할 수 없습니다.
  + 즉, 이 결과는 "해당 비즈니스 지표는 과거의 자기 자신 데이터만으로는 미래 방향성을 뚫어볼 수 없다"는 매우 중요한 통계적 증거가 됨

### 3. 분석 전략의 방향 전환 (Action Item)

- 비즈니스 실무에서 데이터가 순수한 확률보행으로 판명되었다면, 분석가는 전략을 완전히 수정해야 합니다.
  + **내부 변수에서 외부 변수로:** 과거 데이터(AR, MA)를 파고드는 1차원적인 시계열 분석(ARIMA)은 여기서 중단해야 합니다. 대신 날씨, 경쟁사의 가격 정책, 거시경제 지표, 마케팅 예산 등 외부 변수(Exogenous variables)를 결합한 ARIMAX 모델이나 머신러닝 알고리즘으로 분석 방향을 틀어야 합니다.
  + **'점 예측'에서 '리스크 관리'로:** "다음 달 매출액은 정확히 100만 원일 것이다"라는 점 예측(Point Forecast)은 불가능합니다. 대신 누적 분산이 커진다는 점을 활용하여 몬테카를로 시뮬레이션을 돌리고, "최악의 경우 70만 원까지 떨어질 수 있으니 예비 자금을 확보하자"는 식의 **시나리오 및 리스크 관리 전략**으로 경영진에게 보고 필요

- "과거 데이터만 믿고 섣불리 미래를 예측하는 비즈니스적 오판을 통계적으로 막아낸 아주 가치 있는 결론"이라고 해석

## 과거의 평균 활용

```python
mean = np.mean(train.value)
test.loc[:, 'pred_mean'] = mean
test.head()
```

```python
last_value = train.iloc[-1].value
test.loc[:, 'pred_last'] = last_value
test.head()
```

## 표류 기법 활용
- 마지막으로 측정된 값으로 예측하는 방법 수정
- 시간의 흐름에 따른 값의 증가나 감소 허용
- 훈련 집합의 첫번째 값과 마지막 값 사이의 기울기 계산한 뒤, 이 기울기를 미래에 단순히 외삽

```python
deltaX = 800 - 1
deltaY = last_value - 0
drift = deltaY / deltaX
x_vals = np.arange(801, 1001, 1)
pred_drift = drift * x_vals
test.loc[:, 'pred_drift'] = pred_drift
test.head()
```

## 예측 시각화 구현

```python
fig, ax = plt.subplots()

ax.plot(train.value, 'b-')
ax.plot(test['value'], 'b-')
ax.plot(test['pred_mean'], 'r-.', label='평균')
ax.plot(test['pred_last'], 'g--', label='마지막 값')
ax.plot(test['pred_drift'], 'k:', label='표류')

ax.axvspan(800, 1000, color='#808080', alpha=0.2)
ax.legend(loc=2)

ax.set_xlabel('Timesteps')
ax.set_ylabel('Value')

plt.tight_layout()
plt.show()
```

## 예측모형 평가

```python
from sklearn.metrics import mean_squared_error

mse_mean = mean_squared_error(test['value'], test['pred_mean'])
mse_last = mean_squared_error(test['value'], test['pred_last'])
mse_drift = mean_squared_error(test['value'], test['pred_drift'])

print(mse_mean, mse_last, mse_drift)
```
