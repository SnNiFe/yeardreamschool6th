# 문제 2
[문제 2] - 한 은행에서 고객 100명을 대상으로 정기예금 가입 여부를 조사하였다. 각 고객에는 나이(age), 연간 평균 잔액(balance), 최근 마케팅 캠페인 접촉 횟수(campaign), 그리고 정기예금 가입 여부(y)가 포함되어 있다. 로지스틱 회귀 분석을 통해 고객의 정기예금 가입 여부를 예측하고자 한다.  (임곗값 0.5 기준)

[데이터]
- 파일명: 7_3_2_train.csv / 7_3_2_test.csv
- 출처: UCI Machine Learning Repository - Bank Marketing Dataset
    + 원본: Moro et al. (2014), https://archive.ics.uci.edu/dataset/222
    + 링크: https://doi.org/10.24432/C5K306

[Data Description]
※ `7_3_2_train.csv / 7_3_2_test.csv`는 포르투갈 은행의 전화 마케팅 캠페인 데이터에서 무작위로 선택된 100명의 고객 정보를 포함한다. 원본 데이터는 2008년 5월부터 2010년 11월까지 수집되었으며, 정기예금 가입 여부를 예측하는 것이 목표이다.

[컬럼 설명]
- customer_id: 고객 ID (1~100)
- age: 나이 (세)
- balance: 연간 평균 계좌 잔액 (유로)
- campaign: 이번 캠페인에서 고객에게 연락한 횟수 (회)
- y: 정기예금 가입 여부 (0: 미가입, 1: 가입)

[문항]
(1) 로지스틱 회귀 분석을 수행하고, train 데이터를 사용하여 로지스틱 회귀 분석을 수행하고, balance(연간 평균 잔액) 변수의 오즈비를 계산하시오. (소수점 셋째 자리까지 반올림)
- 오즈비는 로지스틱 회귀 분석에서 독립변수가 한 단위 증가할 때, 종속변수가 발생할 오즈(odds)가 몇 배 증가하는지를 나타내는 지표입니다.
- 오즈(Odds): 어떤 사건이 발생할 확률을 발생하지 않을 확률로 나눈 값
- 오즈비(Odds Ratio): 독립변수가 1단위 증가할 때의 오즈 변화율

(2) train 데이터로 구축한 로지스틱 회귀 모형의 **잔차 이탈도(Residual Deviance)**를 계산하시오. (소수점 셋째 자리까지 반올림)
- 잔차 이탈도는 로지스틱 회귀 모형의 적합도를 평가하는 지표로, 모형이 데이터를 얼마나 잘 설명하는지를 나타냅니다.
    + Null Deviance: 독립변수가 없는 기본 모형의 이탈도
    + Deviance Reduction = Null Deviance - Residual Deviance (클수록 모형이 유용)

(3) test 데이터로 오분류율을 계산하시오.
    + test 데이터에 대해 구축한 로지스틱 회귀 모형을 적용하여 **오분류율(Misclassification Rate)**을 계산하시오. (소수점 셋째 자리까지 반올림)
    + 오분류율은 분류 모형의 성능을 평가하는 가장 기본적인 지표로, 전체 예측 중 잘못 분류된 비율을 나타냅니다.
    + 분류 기준: 예측 확률 ≥ 0.5 → 가입(1), < 0.5 → 미가입(0)
