## Anderson–Darling 정규성 검정 핵심 정리

- 목적 : 데이터가 **정규분포를 따른다고 볼 수 있는지** 검정

- 가설  
  + H0 : 정규분포를 따른다  
  + H1 : 정규분포를 따르지 않는다  

- 검정 통계량(stat)  
  + 정규분포에서의 **이탈 정도**를 수치화  
  + 값이 클수록 비정규성 큼  
  + 꼬리(tail) 영역 차이에 민감 → 극단값 탐지에 유리  

- p-value 해석 (α=0.05)  
  + p < 0.05 → H0 기각 → 정규분포 아님  
  + p ≥ 0.05 → H0 기각 못함 → 정규성 가정 가능  

- 주의사항  
  + 표본 많으면 작은 차이도 쉽게 기각  
  + 표본 너무 적으면 검정력 약함  
  + 히스토그램·QQ-plot과 함께 해석 권장  

- 본 문제 적용  
  + 시간별 PM2.5 → 날짜별 평균 → 일평균 100개 생성  
  + 100개 값에 A–D 정규성 검정 수행  
  + `normal_ad(x)` → (stat, p_value) 반환  

---

## 다른 정규성 검정과 비교

- Shapiro–Wilk  
  + 소표본에서 검정력 우수  
  + 대표본에서는 쉽게 기각 가능  

- Kolmogorov–Smirnov  
  + 모수 고정 시 적합  
  + 모수 추정 시 Lilliefors 검정 권장  
  + 꼬리 민감도는 A–D보다 약함  

---

## 요약
- 꼬리 민감도 : Anderson–Darling  
- 소표본 정규성 : Shapiro–Wilk  
- 모수 추정 포함 : Lilliefors(K–S 변형)  
- 표본 100개 수준 : A–D 또는 Shapiro–Wilk 사용 일반적


```python
import numpy as np
import pandas as pd

# Anderson-Darling 정규성 검정(stat, p-value)
from statsmodels.stats.diagnostic import normal_ad

# Colab이면 드라이브 경로로 변경해서 사용 가능
DATA_PATH = "/content/drive/MyDrive/Colab Notebooks/Udemy/빅데이터분석기사_파이썬/작업형_제3유형/6회/data/6_3_1.csv"

print("[데이터 로드]")
df = pd.read_csv(DATA_PATH)
# print("shape:", df.shape)
# print(df.head(), "\n")

# pm25 컬럼을 숫자로 변환
df["pm25"] = pd.to_numeric(df["pm25"], errors="coerce")

# date에서 "날짜(연-월-일)"만 떼어내기
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# date에서 "날짜(연-월-일)만" 떼어내기 (시간 정보 제거 목적)
df["date_only"] = df["date"].dt.date

# print(df.head())

# 날짜별(pm25) 평균을 계산해서 "일평균 데이터" 생성
daily = (
    df.groupby("date_only", as_index=False)["pm25"]
    .mean()
    .dropna()
    .sort_values("date_only")
    .reset_index(drop=True)
)

# print(daily.head())

# 1) 3월 2일 시작 ~ 연속 100일 구간 만들기
base_year = int(df["year"].min())
# print(base_year)

# 시작일
start_dt = pd.to_datetime(f"{base_year}-03-02")

# 종료일
end_dt = start_dt + pd.Timedelta(days=99)
# print("기간 정의")
# print("start:", start_dt.date(), "| end:", end_dt.date(), "\n")


# daily에서 범위 필터링
d = pd.to_datetime(daily["date_only"])
daily_100 = daily[(d.between(start_dt, end_dt))].iloc[:100].copy()
# print(daily_100.head())
# print(daily_100.tail())

# 분석 대상
pm_daily = daily_100["pm25"].astype(float)
print("days_used:", len(pm_daily))

# 평균 및 표준편차 계산
mean_ = float(pm_daily.mean())

# 표본 표준편차
# - ddof=1: n-1로 나누는 표본 표준편차
std_ = float(pm_daily.std(ddof=1))
print(std_)

# 문제 요구에 맞춰서 정리
ans1_mean = round(mean_, 3)
ans1_std = round(std_, 3)

# Anderson-Darling 정규성 검정
ad_stat, p_value = normal_ad(pm_daily.values)

# 문제 요구에 맞춰서 정리
ans2_stat = round(float(ad_stat), 3)
ans3_p = round(float(p_value), 3)

# p_value < 0.05 이면, 귀무가설을 기각
# p_value >= 0.05 이면, 귀무가설을 채택
decision = ["채택", "기각"][float(ans3_p) < 0.05]
print("유의수준 5% 결론(H0: 정규분포) =", decision)
```

# 문제 2

[문제 2] - 한 레스토랑에서 고객들의 팁(tip) 금액에 영향을 미치는 요인을 분석하기 위해 데이터를 수집하였다. 다음은 고객들의 식사 정보를 기록한 데이터이다.

[데이터]
- 파일명: 6_3_2.csv
- 출처: Seaborn Built-in Dataset
  - 원본: Bryant, P.G. and Smith, M (1995) "Practical Data Analysis: Case Studies in Business Statistics"
  - 링크: https://github.com/mwaskom/seaborn-data/blob/master/tips.csv

[Data Description]
※ `6_3_2.csv`는 는 한 레스토랑에서 수집된 244건의 식사 기록을 포함하며, 각 레코드는 고객이 지불한 총 식사 금액, 팁, 성별, 흡연 여부, 요일, 시간대, 일행 수 등의 정보를 담고 있다.

[컬럼 설명]
- total_bill: 총 식사 금액 (달러)
- tip: 팁 금액 (달러) - 종속변수
- sex: 성별 (Male/Female)
- smoker: 흡연 여부 (Yes/No)
- day: 요일 (Thur/Fri/Sat/Sun)
- time: 시간대 (Lunch/Dinner)
- size: 일행 수 (명)


[문항]
(1) 팁 금액(tip)을 종속 변수로 하고, 총 식사 금액(total_bill), 일행 수(size)를 독립 변수로 하는 다중회귀 분석을 수행하여 회귀 계수가 가장 높은 변수를 구하시오. (다중회귀모형 적합 시, 절편 포함)

(2) 유의수준 5% 하에서 각 독립 변수가 팁 금액에 미치는 영향이 통계적으로 유의미한지 판단하고, 유의미한 변수의 개수를 구하시오.

(3) 위에서 구축한 회귀모형의 결정계수(R²)를 구하시오. (소수점 셋째 자리까지 반올림)

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from pathlib import Path

DATA_PATH = "/content/drive/MyDrive/Colab Notebooks/Udemy/빅데이터분석기사_파이썬/작업형_제3유형/6회/data/6_3_2.csv"

# print("[데이터 로드]")
df = pd.read_csv(DATA_PATH)
# print("shape:", df.shape)
# print(df.head(), "\n")

# 회귀모델(OLS) 적합
formula = "tip ~ total_bill + size + C(sex) + C(smoker) + C(day) + C(time)"
model = smf.ols(formula=formula, data=df).fit()

# print("[회귀 요약]")
# print(model.summary(), "\n")

# (1) 회귀계수가 가장 높은 변수 찾기
params = model.params.drop(["Intercept", "const"], errors="ignore")
# print(params)

# - idxmax() 활용 : 최대값을 가지는 인덱스(변수명)을 반환
max_var = params.idxmax()
# print("(1) 회귀 계수가 가장 높은 변수 =", max_var)


# (2) 유의수준 5%에서 유의미한 변수 개수
pvals = model.pvalues.drop(["Intercept", "const"], errors="ignore")
# print(pvals)

# 유의수준 5% 기준으로 유의미한 변수 개수 세기
sig_count = int((pvals < 0.05).sum())
# print(sig_count)

# (3) 결정계수
# - 모델이 종속변수(tip)의 분산을 얼마나 설명하는지 나타내기
r2 = round(float(model.rsquared), 3)
print("(3) 결정계수(R^2) =", r2)
```
