# 문제 1 - 독립표본 t-검정

## [문제 1] 온라인 vs 오프라인 채널 영업 사원 월매출 비교 *(세일즈·마케팅)*

대형 유통기업 세일즈팀에서 온라인 채널과 오프라인 채널 영업 사원을 각각 80명, 75명 무작위 선발하여 **지난달 월매출(만 원)**을 수집하였다.  
두 채널 간 월매출에 유의미한 차이가 있는지 **독립표본 t-검정**으로 분석하고자 한다.

**[데이터]**
- 파일명: `sales_1_ttest_indep.csv`

**[컬럼 설명]**
- `emp_id` : 직원 식별 번호
- `channel` : 채널 구분 (온라인 / 오프라인)
- `monthly_sales` : 월매출 (만 원, 실수형)

**[문항]**
(1) 온라인·오프라인 각각의 월매출 **평균**을 구하시오. (소수점 셋째 자리까지 반올림)
(2) Levene 검정으로 등분산 여부를 판단한 뒤, 적절한 독립표본 t-검정을 실시하여 **검정통계량(t)**과 **p-value**를 구하시오. (소수점 셋째 자리까지 반올림)
(3) 유의수준 5%에서 귀무가설의 **기각/채택** 여부를 결정하시오.

---
## 독립표본 t-검정 핵심 정리

- **목적** : 두 독립 집단의 평균 차이가 통계적으로 유의한지 검정

- **가설**
  - H0 : 두 채널의 월매출 평균은 같다 (mu_online = mu_offline)
  - H1 : 두 채널의 월매출 평균은 다르다

- **등분산 검정 (Levene)**
  - p >= 0.05 : 등분산 가정 → `equal_var=True`
  - p < 0.05  : 이분산 → Welch t-test (`equal_var=False`)

- **p-value 해석 (alpha=0.05)**
  - p < 0.05  → H0 기각 → 두 채널 평균 차이 있음
  - p >= 0.05 → H0 채택 → 유의한 차이 없음

```python
import numpy as np
import pandas as pd
from scipy import stats

df = pd.read_csv(DATA_PATH + "sales_1_ttest_indep.csv", encoding="utf-8-sig")

```
