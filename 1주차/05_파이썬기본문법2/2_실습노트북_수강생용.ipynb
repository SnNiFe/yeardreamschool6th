## 2. `for` 문

```
for 변수 in 시퀀스 :
    <수행할 명령>
```

- 시퀀스(리스트 등)의 원소를 **하나씩** 변수에 넣어가며 명령 실행
- 명령은 `len(시퀀스)` 번 실행됨
- **들여쓰기**로 반복할 범위를 구분!

````python
# 기본 예제: [1, 3, 5] 원소를 순서대로 더하기
sum_val = 0 # 현재 기본값 0

for i in [1, 3, 5, 7, 9, 11, 13, 15] :
  print(sum_val, i) # 중간과정 보여주기
  sum_val += i

print("합계:", sum_val) # 최종결과 보여주기
````

````python
# 리스트 원소 개수 세기 (len() 사용하지 않고 직접!)
length = 0
num_list = [1, 3, 5] # 3개

for i in num_list :
  length += 1

print("길이:", length)
````

## 3. `for-range` 문

### `range()` 함수
- **연속되는 숫자**를 만들어 주는 시퀀스 자료형
- `range(a, b)` → a 이상 **(b-1)** 이하 숫자 생성
- `range(a)` → 0 이상 **(a-1)** 이하 숫자 생성 (a번 반복)

````python
# range 동작 확인
print(range(1, 9))    # range 객체
print(list(range(1, 9)))  # b-1 까지 출력
print(list(range(5)))
````

````python
# a 이상 b 미만의 수를 변수에 넣으며 반복
num_list = [1]

for i in range(2,5) : # range(2, 5) ==> [2, 3, 4]
  num_list.append(i)

print(num_list) # 출력 [1, 2, 3, 4]
````

## 4. `while` 문

```
while 조건 :
    <수행할 명령>
```

- **조건이 True인 동안** 명령을 반복 실행
- 조건이 False가 되면 반복 종료
- while문은 누가 주로 사용하는가?
  + 개발자 : 필수
  + 분석가 : 옵션 (분석 업무 수행 시, 전 사용해본적이 없음)

````python
# 기본 예제: 5에서 1까지 카운트다운
i = 5
while i > 0 :
  print(i)
  i = i - 1
print("Launch!")
````

````python
# 예제: 1부터 4까지 더하기
i = 1
total = 0

while i < 5 : # i가 5 미만인 동안 반복
  total += i
  i += 1

print("1~4 합:", total)
````

### `break` 문
- 반복문을 **강제로 탈출**할 때 사용
- 보통 `if` 조건과 함께 씀

````python
# break 예제: 1~5 출력 후 종료
i = 1
while 1 > 0 :
  print(i)
  if i >= 5 :
    break   # i가 5 이상이 되면 반복문 탈출
  i += 1

print("종료")
````

##  핵심 정리

| 구문 | 용도 | 패턴 |
|------|------|------|
| `for i in 시퀀스` | 원소 순회 | 리스트, 문자열 등 |
| `for i in range(n)` | n번 반복 | 횟수 기반 |
| `for i in range(a,b)` | 구간 반복 | a 이상 b 미만 |
| `while 조건` | 조건 기반 반복 | 조건이 True인 동안 |
| `break` | 반복 탈출 | if 와 조합 |

>  `for` → **횟수/범위가 정해진 경우**,  `while` → **조건이 있는 경우** 주로 사용!

### `list.pop(i)` — 원소 제거 & 반환

### `seq.count(d)` — 원소 개수 세기
- 시퀀스(리스트, 문자열) 안에서 `d`가 몇 번 나오는지 반환

````python
# 리스트에서 count
my_seq = [2, 2, 2, 4, 4]
# 2일 떄, 숫자 1씩 추가
i = 0
for j in my_seq :
  if j == 2 :
    i += 1
print(i) # 반복문으로 셈

print("2의 개수:", my_seq.count(2)) # 개수 3

# 위의 코드는 list 클래스 내부에 있는 count() 매서드

# 문자열에서 count
my_str = "elice and alice"
print("e의 개수:", my_str.count("e")) # 개수 3

# 위의 코드는 str 클래스 내부에 있는 count() 매서드
````

## 3. Dictionary (딕셔너리)

> 사전처럼 **"단어(Key)"** 와 **"뜻(Value)"** 이 짝을 이루는 자료형

- 중괄호 `{}` + 콜론 `:` 으로 표현: `{Key: Value, ...}`
- **Key**는 변할 수 없는 자료형이어야 함 (문자열, 숫자, 튜플 가능 / 리스트 불가)
- `dict[Key]` 로 Value 접근
- `keys()`, `values()`, `items()` 메서드 제공

````python
# 딕셔너리 생성
empty_dict = {} # 빈 딕셔너리 생성
person = {"name" : "Rabbit", "age" : 22}
print(person)
print(type(person)) # <class 'dict'>
````

````python
# 값 접근 & 추가 & 삭제
person = {"name" : "Rabbit", "age" : 22} # key & value

# 접근
print(person["name"]) # Rabbit
print(person["age"]) # 22

# 추가
person["mail"] = "rabbit@elice.com"
print(person)

# 삭제
del person["age"]
print(person)

# 수정
person["name"] = "Kim"
print(person)
````

````python
# keys(), values(), items()

person = {"name": "Rabbit", "age": 22, "mail": "rabbit@elice.com"}
print(person.keys())
print(person.values())
print(person.items())

print(type(person.keys()))
print(list(person.keys()))
````

### ️ 실습 2-1: 내장 함수 종합

## 3. 사용자 정의 함수

```python
def 함수이름(매개변수):
    <수행할 명령>
    ...
    return 반환값
```

- `def` : define(정의하다)의 약자
- **매개변수(parameter)** : 함수 내부로 값을 전달하는 변수
- **들여쓰기**로 함수 내부 범위 구분
- `return` : 함수 외부로 값을 전달 (없으면 `None` 반환)

````python
# 기본 구조 예시
def plusDouble(a, b):
  c = a + b           # 두 수를 더함
  return 2 * c        # 합의 2배를 반환

result = plusDouble(3, 4)
print(result)
````

### `return` 이 왜 필요할까?

````python
# return 없을 때 → 함수 밖에서 결과를 사용할 수 없음
def formula_no_return(a, b):
    c = (a**2) + (b**2)
    print("함수 안:", c)     # 출력은 되지만...

# formula_no_return(2, 4) 이런 류의 코드는 내부 함수가 잘 동작하는 지 확인 용도
result = formula_no_return(2, 4)
print("함수 밖:", result)    # None ← 반환값이 없으면 None!

print()   # 그냥 줄 넘기기

# return 있을 때 → 결과를 변수에 저장 가능
def formula_with_return(a, b):
    c = (a**2) + (b**2)
    return c                 # 값을 반환!

result = formula_with_return(2, 4)
print("함수 밖:", result)    # 20 ← 정상적으로 값을 받음
````

# 함수 Docstring & 타입 힌트 (Type Hints)
- 파이썬 함수를 전문적으로 작성하는 두 가지 핵심 관행으로, 코드의 가독성과 유지보수성을 높입니다.

## 타입 힌트

````python
# 타입 힌트 없음 → 매개변수와 반환값의 자료형을 알 수 없음
def add(a, b):
    return a + b


# 타입 힌트 있음 → int 두 개를 받아 int를 반환한다는 것이 명확
def add(a: int, b: int) -> int:
    return a + b
````

````python
# 다양한 자료형 예시
def greet(name: str) -> str:           # str → str
    return "안녕하세요, " + name

def is_even(n: int) -> bool:            # int → bool
    return n % 2 == 0

def average(nums: list) -> float:      # list → float
    return sum(nums) / len(nums)

def log_message(msg: str) -> None:    # 반환값 없으면 None
    print("[LOG]", msg)
````

## Docstring — 함수 설명 문서
- 목적: 함수의 역할, 매개변수, 반환값을 자연어로 설명
- help(함수명) 또는 IDE에서 함수 위에 커서를 올리면 docstring 내용이 자동으로 표시됩니다. 오픈소스 라이브러리의 공식 문서도 대부분 docstring에서 자동 생성됩니다.

````python
def celsius_to_fahrenheit(celsius: float) -> float:
    """섭씨 온도를 화씨 온도로 변환합니다.

    Args:
        celsius (float): 변환할 섭씨 온도 값

    Returns:
        float: 변환된 화씨 온도 값

    Example:
        >>> celsius_to_fahrenheit(100)
        212.0
    """
    return celsius * 9 / 5 + 32
````

##  타입 힌트 + Docstring 완전체 예시
- 실무에서 함수를 이렇게 작성합니다

````python
def get_grade(score: int) -> str:
    """점수를 받아 학점 문자열을 반환합니다.

    Args:
        score (int): 0~100 사이의 시험 점수

    Returns:
        str: 학점 문자열 ("A", "B", "C", "D", "F")

    Raises:
        ValueError: score가 0~100 범위를 벗어난 경우

    Example:
        >>> get_grade(95)
        'A'
        >>> get_grade(72)
        'C'
    """
    if not (0 <= score <= 100):
        raise ValueError(f"score는 0~100이어야 합니다. 입력값: {score}")
    if score >= 90:   return "A"
    elif score >= 80: return "B"
    elif score >= 70: return "C"
    elif score >= 60: return "D"
    else:             return "F"
````

````python
result = get_grade(85)
print(result)        # B
````

````python
help(get_grade)      # docstring 내용이 출력됨
````

````python
# 연습: 다음 조건을 모두 갖춘 함수를 작성하세요
# - 타입 힌트 (매개변수 + 반환값)
# - docstring (역할 / Args / Returns / Example)
# - PEP 8 스타일 준수

def calculate_bmi(weight: float, height_cm: float) -> float:
    """체중(kg)과 신장(cm)으로 BMI를 계산합니다.

    Args:
        weight (float): 체중 (단위: kg)
        height_cm (float): 신장 (단위: cm)

    Returns:
        float: BMI 값 (소수점 둘째 자리 반올림)

    Example:
        >>> calculate_bmi(70, 175)
        22.86
    """
    height_m = height_cm / 100
    bmi = weight / (height_m ** 2)
    return round(bmi, 2)


# 실행 결과 확인
print(calculate_bmi(70, 175))  # 22.86
print(calculate_bmi(55, 162))  # 20.96
help(calculate_bmi)            # docstring 확인
````
