## 1.5 문자열·리스트 접근하기 — 인덱싱 & 슬라이싱

- **인덱스:** 원소의 위치 번호, **0부터 시작**, 음수 인덱스는 뒤에서부터
- **인덱싱:** `자료[인덱스]` → 특정 위치의 원소 하나 반환
- **슬라이싱:** `자료[시작:끝]` → 시작 이상, 끝 **미만** 범위 반환
- 끝을 생략하면 마지막까지, 시작을 생략하면 처음부터

```
"Ready"  →  R  e  a  d  y
인덱스   →  0  1  2  3  4  (양수)
           -5 -4 -3 -2 -1  (음수)
```


````python
word    = "Ready"
numbers = [2, 4, 6, 8, 10]

# 인덱싱
print(word[0]) # R --> 첫번째 값
print(word[1]) # e --> 두번째 값

# numbers 리스트에서 숫자 6을 추출하기
print(numbers[2])
print(numbers[-3]) # 음수 거꾸로 추출하기, -1부터 시작

# 슬라이싱
#       0  1  2  3  4   5   6
beta = [2, 4, 6, 8, 10, 12, 14]
print(beta[0:3]) # 인덱스 0 이상 3 미만
print(beta[:3]) # 인덱스 0 이상 3 미만
print(beta[2:]) # 인덱스 2 이상 부터 끝까지
print(beta[:]) # 인덱스 전체 출력

sentence = "Hi! elice!"
````

## 2.2 논리형 & 비교·논리 연산자

| 연산자 | 의미 | 예시 | 결과 |
|--------|------|------|------|
| `==` | 같다 | `3 == 3` | `True` |
| `!=` | 다르다 | `3 != 5` | `True` |
| `<` `>` | 크기 비교 | `3 < 5` | `True` |
| `<=` `>=` | 이하/이상 | `3 >= 3` | `True` |
| `and` | 모두 True | `True and False` | `False` |
| `or` | 하나라도 True | `True or False` | `True` |
| `not` | 반전 | `not True` | `False` |


````python
# 비교 연산자
print(3 < 5)    # True
print(7 == 5)   # False
print(2 >= 10)  # False
print(5 != 10)  # True

# and: 모두 True여야 True
print(3==3 and 4<=5 and 6>2)   # True
print(3==3 and 4>5)            # False

# or: 하나라도 True면 True
print(3==4 or 4<=5 or 6<2)    # True
print(3==4 or 5==6)            # False

# not: 논리값 반전
print(not 3==4)   # True   (False → True)
print(not 3==3)   # False  (True → False)
````

# 3. 리스트

## 핵심 개념
- **리스트 생성:** `[]` 또는 `list()` — 빈 리스트부터 혼합 자료형까지
- **`append(d)`:** 맨 끝에 원소 하나 추가
- **`insert(i, d)`:** 인덱스 `i` 위치에 원소 삽입, 기존 원소는 뒤로 밀림
- **`remove(d)`:** 첫 번째로 등장하는 `d` 삭제 (없으면 ValueError)
- **`sort()`:** 원본 리스트를 오름차순 정렬, `reverse=True` 로 내림차순
- **시퀀스 자료형 공통 특징:** 인덱싱 · 슬라이싱 · `in` · `len()` · `+` · `*`
- 문자열은 `append` 불가 / 리스트는 문자 기준 `split` 불가 — 종류별 차이 주의


## 3.1 리스트 핵심 메서드

- 클래스와 메서드를 공부할 때, 리스트 클래스와 리스트 클래스 메서드를 기억하자!

- 반복문 사용할 때, 빈번하게 사용 됨

## 3.2 시퀀스 자료형 공통 특징

- **순서가 있는** 자료형 — 문자열(`str`)과 리스트(`list`) 모두 해당
- 공통 기능: 인덱싱, 슬라이싱, `in`, `not in`, `len()`, `+`, `*`
- 종류별 차이: `list.append()` 가능 ↔ `str.append()` 불가

- 에러메세지
  + AttributeError: 'str' object has no attribute 'append'
  + str 객체는 'append'라는 메서드가 존재하지 않는다.


````python
a = "Hello"
b = ["e", "l", "i", "c", "e"]

b.append("a")
# b.Append("a") # list 객체는 Append 라는 메서드가 존재하지 않는다 (AttributeError: 'list' object has no attribute 'Append')
# a.append("!") # srt 객체는 append 라는 메서드가 존재하지 않는다 (AttributeError: 'str' object has no attribute 'append')
````

````python
a = "Hello"
b = ["e", "l", "i", "c", "e"]

# 인덱싱 & 슬라이싱
print(a[1])      # e
print(b[2:4])    # ['i', 'c']
print(a[-1])     # o
print(b[:4])     # ['e', 'l', 'i', 'c']

# in 연산자
print("e" in a)     # True
print("E" in b)     # False (대소문자 구분)
print("z" not in a) # True

# len()
print(len("Hello World"))           # 11 (공백 포함)
print(len(["e","l","i","c","e"]))   # 5

# + 이어붙이기
print("Good" + "Day")                       # GoodDay
print(["H","e"] + ["l","l","o"])            # ['H','e','l','l','o']

# * 반복
print("Go!!" * 5)                           # Go!!Go!!Go!!Go!!Go!!
print([13, 17, 19] * 3)                     # [13,17,19, ...]
````

## 시나리오 1 : 학생 성적 관리 시스템

### 목표
- 학생 이름과 점수를 리스트로 관리
- 인덱싱으로 특정 학생 조회
- 조건문으로 합격/불합격 판별
- `sort()` 로 성적 순위 확인

### 사용 개념
- 리스트 생성 · 인덱싱 · 슬라이싱
- `for` 반복문 (미리 보기)
- `if-else` 조건문
- `sort()` · `len()`


````python
# 해보기
name = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
score = [85, 72, 90, 65, 88]
num = len(name)
print("학생 이름:", name)
print("학생 성적:", score)
print("총 학생 수:", num)
print("")

person = ""
person = str(input("조회 학생: "))
if person == name[0] or person == name[1] or person == name[2] or person == name[3] or person == name[4] :
  p_index = name.index(person)
  p_score = score[p_index]
  print("해당 성적:", p_score)
else :
  person = name[0]
  print("해당 성적: error")
print("")

min = 70 # 합격 기준
print("[ 합격 기준: ", min, "점 ]")
i = 0
corr = False
if score[i] >= min :
  print(name[i],"   " , score[i], "점 ->", "합격")
else :
  print(name[i],"   " , score[i], "점 ->", "불합격")
i = 1
corr = False
if score[i] >= min :
  print(name[i],"   " , score[i], "점 ->", "합격")
else :
  print(name[i],"   " , score[i], "점 ->", "불합격")
i = 2
corr = False
if score[i] >= min :
  print(name[i],"   " , score[i], "점 ->", "합격")
else :
  print(name[i],"   " , score[i], "점 ->", "불합격")
i = 3
corr = False
if score[i] >= min :
  print(name[i],"   " , score[i], "점 ->", "합격")
else :
  print(name[i],"   " , score[i], "점 ->", "불합격")
i = 4
corr = False
if score[i] >= min :
  print(name[i],"   " , score[i], "점 ->", "합격")
else :
  print(name[i],"   " , score[i], "점 ->", "불합격")
print("")

h_score = score
h_score.sort(reverse=True)
print("정렬된 성적:", h_score)
print("최저:", h_score[4], " / 최고:", h_score[0])
print("상위 3개 성적:", h_score[0:3])
print("")

n_name = name
n_name.append("Frank")
n_score = score
n_score.append(int(78))
print("추가 후 명단:", n_name)
print("추가 후 성적:", n_score)
n_score_r = n_score
n_score_r.remove(65)
print("65점 제거 후:", n_score_r)
````
